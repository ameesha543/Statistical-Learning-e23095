# Triangular Distribution Analysis: Generator Installation

Given parameters:
- Minimum ($a$) = 4
- Maximum ($b$) = 10
- Mode ($c$) = 6

### 1. Support of $X$
The support is the interval where the density is non-zero:
<br>**$[4, 10]$**.

### 2. Probability Density Function $f_X(x)$
The PDF for a Triangular distribution is:
$$
f_X(x) = \begin{cases}
\frac{2(x-a)}{(b-a)(c-a)} & a \leq x < c \\
\frac{2(b-x)}{(b-a)(b-c)} & c \leq x \leq b \\
0 & \text{otherwise}
\end{cases}
$$
Plugging in $a=4, b=10, c=6$:
$$
f_X(x) = \begin{cases}
\frac{2(x-4)}{(10-4)(6-4)} = \frac{2(x-4)}{12} = \frac{x-4}{6} & 4 \leq x < 6 \\
\frac{2(10-x)}{(10-4)(10-6)} = \frac{2(10-x)}{24} = \frac{10-x}{12} & 6 \leq x \leq 10 \\
0 & \text{otherwise}
\end{cases}
$$

```python
import scipy.stats as stats
import numpy as np
from scipy.integrate import quad   # ← This was the missing import

a, b, c = 4, 10, 6

# Scipy uses (c-a)/(b-a)
scipy_c = (c - a) / (b - a)
loc = a
scale = b - a
dist = stats.triang(c=scipy_c, loc=loc, scale=scale)

# 3. Verify integration
area, error = quad(lambda x: dist.pdf(x), a, b)
print(f"3. Integral over support: {area}")

# 4. Expectation
mean = dist.mean()
print(f"4. E[X]: {mean:.4f} (Formula: (a+b+c)/3 = {(a+b+c)/3:.4f})")

# 5. Variance
variance = dist.var()
print(f"5. Var(X): {variance:.4f} (Formula: (a²+b²+c²-ab-ac-bc)/18 = {(a**2 + b**2 + c**2 - a*b - a*c - b*c)/18:.4f})")
```

### 6. Interpretation
- **$a=4$**: The absolute minimum time required; unlikely to finish sooner due to physical constraints.
- **$b=10$**: The maximum time; represents the worst-case scenario (e.g., equipment delays).
- **$c=6$**: The most frequent or typical completion time (the peak of the distribution).

### 7. Why use a Triangular Distribution?
It is ideal when data is scarce but expert opinions on bounds and mode are available. It is simpler than a Beta distribution but more descriptive than a Uniform distribution because it considers for a central tendency.



```python
# 8. P(X <= 5)
p_le_5 = dist.cdf(5)
print(f"8. P(X <= 5): {p_le_5:.4f}")

# 9. P(5 <= X <= 8)
p_5_8 = dist.cdf(8) - dist.cdf(5)
print(f"9. P(5 <= X <= 8): {p_5_8:.4f}")
```
