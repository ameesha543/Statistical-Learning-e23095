### 1. Probability Space

A suitable probability space $(\Omega, \mathfrak{F}, P)$ is:
* $\Omega = \{ x \in \mathbb{R} \mid 0 \le x \le 1 \}$, representing all possible proportions of defective items.
* $\mathfrak{F}$: The Borel $\sigma$-algebra on $[0, 1]$, denoted $\mathcal{B}([0, 1])$.
* $P$: Defined by the density function $f(x) = \frac{1}{B(2,18)}x(1-x)^{17}$ for $x \in [0, 1]$, such that for any $A \in \mathfrak{F}$, $P(A) = \int_A f(x) dx$.

### 2. Random Variable $X$
$X: \Omega \to [0, 1]$ is the identity function $X(\omega) = \omega$.<br> Since $\Omega = [0, 1]$, $X$ maps each outcome (the proportion) to itself.

### 3. PDF
As given:
$$f(x) = \frac{x(1-x)^{17}}{B(2, 18)}, \quad 0 < x < 1$$
Note: $B(2, 18) = \frac{\Gamma(2)\Gamma(18)}{\Gamma(20)} = \frac{1! \cdot 17!}{19!} = \frac{1}{19 \times 18} = \frac{1}{342}$

```python
import numpy as np
from scipy.stats import beta
from scipy.special import beta as beta_func

alpha, b_param = 2, 18

# 4. Verify PDF integrates to 1
integral_val = beta_func(alpha, b_param) * 342
print(f"Verification: B(2,18) is {beta_func(alpha, b_param):.6f}, which is 1/342 ({1/342:.6f})")

# 5.  E[X]
expected_val = alpha / (alpha + b_param)

# 6.Var(X)
variance_val = (alpha * b_param) / ((alpha + b_param)**2 * (alpha + b_param + 1))

print(f"E[X]: {expected_val}")
print(f"Var(X): {variance_val:.6f}")
```

### $7.\ P(X \le 0.10)$
In integral form:
$$P(X \le 0.10) = \int_{0}^{0.10} \frac{x(1-x)^{17}}{B(2, 18)} dx = I_{0.10}(2, 18)$$
where $I_x(\alpha, \beta)$ is the regularized incomplete beta function.

### 8. $P(0.05 \le X \le 0.15)$
In integral form:
$$P(0.05 \le X \le 0.15) = \int_{0.05}^{0.15} \frac{x(1-x)^{17}}{B(2, 18)} dx = I_{0.15}(2, 18) - I_{0.05}(2, 18)$$

### 9. Why Beta distribution?
The Beta distribution is ideal for modeling proportions or percentages because its domain is strictly $[0, 1]$. By adjusting $\alpha$ and $\beta$, it can model various shapes (e.g.:skewed, symmetric, U-shaped), making it flexible for manufacturing defects where proportions are usually small.

### 10. Interpretation
* **Expectation :** On average, 10% of items in a batch are defective.
* **Variance :** This measures the consistency of the manufacturing process. A low variance means most batches will have a defective proportion very close to 10%.

```python
# Solutions for 7 and 8
p_le_010 = beta.cdf(0.10, alpha, b_param)
p_range = beta.cdf(0.15, alpha, b_param) - beta.cdf(0.05, alpha, b_param)

print(f"P(X <= 0.10): {p_le_010:.4f}")
print(f"P(0.05 <= X <= 0.15): {p_range:.4f}")
```
