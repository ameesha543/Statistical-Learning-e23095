### PERT Distribution Calculation

Given:
- $a = 5$ (minimum)
- $c = 11$ (most likely)
- $E[X] = c$ (expected value equals mode)

**1. Determine the maximum time $b$**

The formula for the mean of a PERT distribution is:
$$E[X] = \frac{a + 4c + b}{6}$$

Setting $E[X] = c$:
$$c = \frac{a + 4c + b}{6}$$
$$6c = a + 4c + b$$
$$2c = a + b$$
$$b = 2c - a$$
$$b = 2(11) - 5 = 17$$

**2. Variance**

The variance is given by:
$$\sigma^2 = \frac{(E[X] - a)(b - E[X])}{7}$$
Since $E[X] = c = 11$:
$$\sigma^2 = \frac{(11 - 5)(17 - 11)}{7} = \frac{6 \times 6}{7} = \frac{36}{7} \approx 5.14$$

```python
a = 5
c = 11

# Calculate b
b = 2*c - a
print(f'Maximum time b: {b}')

# Calculate expected value to verify
expected_value = (a + 4*c + b) / 6
print(f'Expected Value: {expected_value}')

# Calculate Variance
variance = ((expected_value - a) * (b - expected_value)) / 7
print(f'Variance: {variance:.4f}')
```
