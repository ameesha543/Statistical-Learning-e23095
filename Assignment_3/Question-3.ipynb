## 1. Joint Distribution

### Part A: Distribution of $Y$
The measurement is given by $Y = X + Q$. Since $X$ and $Q$ are independent normal random variables, their sum is also normally distributed.
* **Mean:** $E[Y] = E[X + Q] = E[X] + E[Q] = 0 + 0 = 0$
* **Variance:** Since $X$ and $Q$ are independent, $\text{Var}(Y) = \text{Var}(X) + \text{Var}(Q) = 1 + 4 = 5$

Thus, the marginal distribution of $Y$ is:
$$Y \sim \mathscr{N}(0, 5)$$

### Part B: Covariance $\text{Cov}(X, Y)$
Using the linearity of covariance and the independence of $X$ and $Q$:
$$\text{Cov}(X, Y) = \text{Cov}(X, X + Q) = \text{Cov}(X, X) + \text{Cov}(X, Q)$$
$$\text{Cov}(X, Y) = \text{Var}(X) + 0 = 1$$

### Part C: Joint Distribution Matrix
Since any linear combination of $X$ and $Y$ is a linear combination of the independent normals $X$ and $Q$, the vector $\mathbf{Z} = [X, Y]^T$ has a joint multivariate normal distribution:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathscr{N} \left( \begin{bmatrix} 0 \\ 0 \end{bmatrix}, \begin{bmatrix} 1 & 1 \\ 1 & 5 \end{bmatrix} \right)$$

## 2. Information Geometry

A random variable $X$ is $\sigma(Y)$-measurable if and only if $X$ can be written strictly as a deterministic function of $Y$ (i.e., $X = g(Y)$).

In this system, $X$ is **not** $\sigma(Y)$-measurable because the mapping from $X$ to $Y$ is corrupted by an independent source of randomness, the mechanical load $Q$. For any observed value $Y = y$, there exists an infinite number of pairs $(x, q)$ such that $x + q = y$.

**Practical Interpretation:**
If the power meter reads $Y = 5$, the engineer faces an **identifiability problem**. The high reading could be caused by an overheating motor ($X = 4, Q = 1$) or a completely cool motor cutting extremely hard material ($X = 0, Q = 5$). Because $\sigma(Y)$ only contains information about the *sum*, it is impossible to perfectly disentangle the thermal signal from the mechanical noise using a single power measurement.

## 3. Conditional Expectation (The Best Estimate)

Using the property for joint multivariate normal distributions:
$$E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$$

Substituting our known parameters ($\mu_X = 0, \mu_Y = 0, \text{Cov}(X,Y) = 1, \text{Var}(Y) = 5$):
$$E[X \mid Y = y] = 0 + \frac{1}{5}(y - 0) = \frac{1}{5}y$$

### Signal-to-Total-Variance Ratio
We can express the scaling factor as a ratio of variances:
$$\frac{\text{Cov}(X,Y)}{\text{Var}(Y)} = \frac{\text{Var}(X)}{\text{Var}(X) + \text{Var}(Q)} = \frac{1}{1 + 4} = \frac{1}{5}$$

This demonstrates that our best estimate $\widehat{X} = \frac{1}{5}Y$ scales the observation by the proportion of total variance that is driven by the true underlying signal ($X$).

## 4. Verification of the Residual

Let our estimator be $\widehat{X} = \frac{1}{5}Y$. The residual error is defined as:
$$Z = X - \widehat{X} = X - \frac{1}{5}Y$$

### Part A: Mean of the Residual
$$E[Z] = E\left[X - \frac{1}{5}Y\right] = E[X] - \frac{1}{5}E[Y] = 0 - 0 = 0$$

### Part B: Orthogonality Principle ($\text{Cov}(Z, Y) = 0$)
$$\text{Cov}(Z, Y) = \text{Cov}\left(X - \frac{1}{5}Y, Y\right) = \text{Cov}(X, Y) - \frac{1}{5}\text{Cov}(Y, Y)$$
$$\text{Cov}(Z, Y) = \text{Cov}(X, Y) - \frac{1}{5}\text{Var}(Y)$$

Substituting the known values ($\text{Cov}(X,Y) = 1$ and $\text{Var}(Y) = 5$):
$$\text{Cov}(Z, Y) = 1 - \frac{1}{5}(5) = 1 - 1 = 0$$

Since the error is uncorrelated with our observation, the linear estimator has extracted all accessible information from $Y$.

## 5. Information Theory: Differential Entropy

The differential entropy of a univariate normal variable $W \sim \mathscr{N}(\mu, \sigma^2)$ is given by:
$$h(W) = \frac{1}{2} \ln(2\pi e \sigma^2)$$

### Part A: Differential Entropy $h(X)$
For $X \sim \mathscr{N}(0, 1)$, where $\sigma_X^2 = 1$:
$$h(X) = \frac{1}{2} \ln(2\pi e \cdot 1) = \frac{1}{2} \ln(2\pi e) \approx 1.4189 \text{ nats}$$

### Part B: Conditional Differential Entropy $h(X \mid Y)$
First, we find the conditional variance $\sigma_{X|Y}^2$:
$$\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)} = 1 - \frac{1^2}{5} = 1 - 0.2 = 0.8$$

Since the conditional distribution $(X \mid Y=y)$ is normal and its variance is independent of the specific value of $y$, the conditional entropy is:
$$h(X \mid Y) = \frac{1}{2} \ln(2\pi e \sigma_{X|Y}^2) = \frac{1}{2} \ln(2\pi e \cdot 0.8) \approx 1.3073 \text{ nats}$$

## 6. Mutual Information

The mutual information measures the reduction in uncertainty about $X$ due to observing $Y$:
$$I(X; Y) = h(X) - h(X \mid Y)$$
$$I(X; Y) = \frac{1}{2} \ln(2\pi e) - \frac{1}{2} \ln(2\pi e \cdot 0.8) = \frac{1}{2} \ln\left(\frac{1}{0.8}\right) = \frac{1}{2} \ln(1.25) \approx 0.1116 \text{ nats}$$

### Part A: Percentage of Uncertainty Removed
In information theory, the fraction of uncertainty removed can be quantified by the variance reduction or the normalized mutual information. Looking directly at the **variance reduction**:
$$\text{Uncertainty Reduction} = \frac{\sigma_X^2 - \sigma_{X|Y}^2}{\sigma_X^2} = \frac{1 - 0.8}{1} = 20\%$$

Observing the power meter eliminates **20%** of the prior variance of the internal temperature.

### Part B: Effect of High Noise ($\text{Var}(Q) = 100$)
If mechanical noise increases such that $\text{Var}(Q) = 100$:
* New total variance: $\text{Var}(Y) = \text{Var}(X) + \text{Var}(Q) = 1 + 100 = 101$
* New conditional variance: $\sigma_{X|Y}^2 = 1 - \frac{1^2}{101} = \frac{100}{101} \approx 0.9901$
* New Uncertainty Reduction: $\frac{1 - 0.9901}{1} \approx 0.99\%$

If the environment becomes that noisy, the power meter becomes virtually useless, removing less than **1%** of the temperature uncertainty.

## 7. Diagnostic Interpretation

The manager's claim that *"our best estimate of temperature should just be $Y$ itself"* is incorrect for two primary reasons:

1. **Bias and Overestimation:** If we use $\widehat{X}_{\text{manager}} = Y$, the expected error given a specific temperature $X$ is:
   $$E[\widehat{X}_{\text{manager}} - X \mid X] = E[X + Q - X \mid X] = E[Q] = 0$$
   However, looking at it from a conditional tracking perspective given our observation $Y$, $Y$ systematically overstates the magnitude of $X$ because it includes $Q$. On average, $Y$ is 5 times larger than the true temperature component ($E[X|Y] = \frac{1}{5}Y$).

2. **The Principle of Shrinkage (Bayesian Regularization):**
   The optimal estimator $E[X \mid Y] = \frac{1}{5}Y$ consciously "shrinks" the raw measurement toward the prior mean (zero). Because the power meter is corrupted by significant background noise ($\text{Var}(Q) = 4$), a high reading of $Y$ is statistically more likely to be caused by a temporary spike in mechanical load than by extreme internal temperatures.
   
   The estimator balances the raw data against prior knowledge: when noise is high, it heavily discounts the measurement $Y$ to avoid triggering false alarms for the factory.
