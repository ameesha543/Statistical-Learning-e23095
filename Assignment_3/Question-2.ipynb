## Derivation of the Conditional Multivariate Normal Distribution

To find the conditional density $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$, we use the definition of conditional probability:

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}$$

We can evaluate this by decomposing both the **determinant** and the **quadratic form in the exponent** of the joint density function using the properties of the **Schur complement**.

---

### 1. Matrix Inversion and the Schur Complement

Let the conditional covariance matrix (the Schur complement of $\boldsymbol{\Sigma}_{YY}$ in $\boldsymbol{\Sigma}$) be defined as:
$$\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$$

Using the block matrix inversion formula, the inverse of the joint covariance matrix $\boldsymbol{\Sigma}^{-1}$ can be written as:
$$\boldsymbol{\Sigma}^{-1} = \begin{bmatrix} \boldsymbol{\Sigma}_{X\mid Y}^{-1} & -\boldsymbol{\Sigma}_{X\mid Y}^{-1}\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \\ -\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}\boldsymbol{\Sigma}_{X\mid Y}^{-1} & \boldsymbol{\Sigma}_{YY}^{-1} + \boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}\boldsymbol{\Sigma}_{X\mid Y}^{-1}\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \end{bmatrix}$$

### 2. Decomposition of the Quadratic Form

Let $\mathbf{x}_0 = \mathbf{x} - \boldsymbol{\mu}_X$ and $\mathbf{y}_0 = \mathbf{y} - \boldsymbol{\mu}_Y$. The joint quadratic form in the exponent is given by:
$$Q = \begin{bmatrix} \mathbf{x}_0 \\ \mathbf{y}_0 \end{bmatrix}^T \boldsymbol{\Sigma}^{-1} \begin{bmatrix} \mathbf{x}_0 \\ \mathbf{y}_0 \end{bmatrix}$$

Substituting the block inversion form of $\boldsymbol{\Sigma}^{-1}$ into $Q$ expands to:
$$Q = \mathbf{x}_0^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} \mathbf{x}_0 - \mathbf{x}_0^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}_0 - \mathbf{y}_0^T \boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}\boldsymbol{\Sigma}_{X\mid Y}^{-1}\mathbf{x}_0 + \mathbf{y}_0^T \left( \boldsymbol{\Sigma}_{YY}^{-1} + \boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}\boldsymbol{\Sigma}_{X\mid Y}^{-1}\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \right) \mathbf{y}_0$$

Grouping the terms containing $\boldsymbol{\Sigma}_{X\mid Y}^{-1}$ reveals a perfect square form:
$$Q = \left( \mathbf{x}_0 - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}_0 \right)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} \left( \mathbf{x}_0 - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}_0 \right) + \mathbf{y}_0^T \boldsymbol{\Sigma}_{YY}^{-1} \mathbf{y}_0$$

Now, substituting $\mathbf{x}_0$ and $\mathbf{y}_0$ back into the first term yields:
$$\mathbf{x}_0 - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}_0 = (\mathbf{x} - \boldsymbol{\mu}_X) - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y) = \mathbf{x} - \boldsymbol{\mu}^*$$

Where the conditional mean $\boldsymbol{\mu}^*$ is defined as:
$$\boldsymbol{\mu}^* = \boldsymbol{\mu}_X + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)$$

Thus, the total quadratic form simplifies cleanly to:
$$Q = (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) + (\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y)$$

---

### 3. Decomposition of the Determinant

Using the determinant properties of partitioned matrices, the joint determinant can be factored via the Schur complement as:
$$|\boldsymbol{\Sigma}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\boldsymbol{\Sigma}_{X\mid Y}|$$

---

### 4. Combining into the Conditional Density

Substituting the factored determinant and the simplified quadratic form back into the joint multivariate normal density function gives:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}_{X\mid Y}| |\boldsymbol{\Sigma}_{YY}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) - \frac{1}{2} (\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y) \right)$$

This expression splits cleanly into two distinct multiplying components:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y}) = \left[ \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right) \right] \times \left[ \frac{1}{\sqrt{(2\pi)^m |\boldsymbol{\Sigma}_{YY}|}} \exp\left( -\frac{1}{2} (\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y) \right) \right]$$

The right-hand component is precisely the marginal distribution of $\mathbf{Y}$, denoted as $f_{\mathbf{Y}}(\mathbf{y}) \sim \mathcal{N}(\boldsymbol{\mu}_Y, \boldsymbol{\Sigma}_{YY})$.

Consequently, we can write:
$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y}) = f_{\mathbf{X}\mid \mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) \cdot f_{\mathbf{Y}}(\mathbf{y})$$

Dividing out the marginal density $f_{\mathbf{Y}}(\mathbf{y})$ isolates our target conditional distribution:

$$f_{\mathbf{X}\mid \mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right)$$
