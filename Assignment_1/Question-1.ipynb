disk radius $R = 10$

event $A$ = point lies within $1$ unit of the boundary

 The point's distance $r$ from the center satisfies $R-1 \le r \le R$,
<br>
which is $9 \le r \le 10$.

```python
R = 10
boundary_thickness = 1

r_inner = R - boundary_thickness

print(f"disk radius (R): {R}")
print(f"boundary thickness: {boundary_thickness}")
print(f"r_inner: {r_inner}")
```

### Part 1: Point uniformly distributed with respect to area

If the point is uniformly distributed on the disk,
<br>
the probability of an event is proportional to its covered area.

- Total area of the disk: $Area_{total} = \pi R^2$
- Area of event A (annulus): $Area_A = \pi R^2 - \pi (R-1)^2$

$P(A) = \frac{Area_A}{Area_{total}} = \frac{\pi R^2 - \pi (R-1)^2}{\pi R^2} = \frac{R^2 - (R-1)^2}{R^2}$

```python
area_total = R**2
area_A = R**2 - r_inner**2

p_A_area = area_A / area_total


print(f"P(A) for uniform area distribution: {p_A_area:.4f}")
```

### Part 2: Distance to the point from the center chosen uniformly from $[0,10]$ and direction uniformly from $[0,2\pi)$

The radial distance $r$ is uniformly distributed over $[0, R]$
<br>
The probability density function for $r$ is $f(r) = \frac{1}{R}$ for $0 \le r \le R$

Event A occurs when $r$ is between $R-1$ and $R$, i.e., $9 \le r \le 10$.

$P(A) = \int_{R-1}^{R} f(r) dr = \int_{9}^{10} \frac{1}{10} dr$

```python
p_A_radial = (R - r_inner) / R

print(f"P(A) for uniform radial distribution: {p_A_radial:.4f}")
```

### Part 3: Explanation for the difference

The two probabilities differ because the underlying probability models for choosing the point are different.

1. **Uniformly distributed with respect to area:**  
   This model means that any given unit of area within the disk has an equal chance of containing the chosen point. Because the area of an annulus increases with its radius ($dA = 2\pi r \, dr$), points further from the center are more likely to be chosen than points closer to the center. The outer regions, therefore, get proportionally more weight.

2. **Distance to the point from the center chosen uniformly from $[0, 10]$:**  
   This model implies that the radial distance $r$ itself is uniformly distributed. This gives equal probability to any interval of $r$ of the same length (e.g., $[0,1]$ has the same probability as $[9,10]$). However, when considering the actual *area* covered by these radial intervals, a small interval near the center (e.g., $r \in [0,1]$) covers a much smaller area than a similar interval near the boundary (e.g., $r \in [9,10]$). This effectively 'overweights' the central region and 'underweights' the boundary region compared to the uniform area distribution.

In essence, a uniform distribution of radial distance does **not** lead to a uniform distribution of area. A uniform area distribution concentrates more probability density *per unit of radial length* at larger radii, whereas a uniform radial distribution gives equal likelihood to all radial intervals, irrespective of the area they encompass.

Therefore, the probability of being in the outer 1-unit band is higher when the area is uniformly distributed because that band contains a larger proportion of the total area.
