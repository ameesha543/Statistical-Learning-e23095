### Answers to the Questions

#### 1. The Probability Space $(\Omega, \mathfrak{F}, P)$

- **Sample Space**: $\Omega = \{G_1, G_2, D_1, D_2\}$. These are the four possible outcomes of inspecting a product.

- **$\sigma$-algebra**: $\mathfrak{F} = \mathcal{P}(\Omega)$, which is the Power Set of $\Omega$. Since $\Omega$ has 4 elements, $\mathfrak{F}$ contains $2^4 = 16$ possible subsets (events).  
  For example:  
  $\mathfrak{F} = \{\emptyset, \{G_1\}, \{G_2\}, \{D_1\}, \{D_2\}, \{G_1, G_2\}, \{G_1, D_1\}, \dots, \Omega\}$.

- **Probability Measure $P$**: Defined on the singletons of $\Omega$ as given:  
  $P(\{G_1\}) = 0.50$  
  $P(\{G_2\}) = 0.20$  
  $P(\{D_1\}) = 0.10$  
  $P(\{D_2\}) = 0.20$

The probability of any other event $A \in \mathfrak{F}$ can be found by summing the probabilities of the constituent singletons.  
For example:  
$P(\{G_1, G_2\}) = P(\{G_1\}) + P(\{G_2\}) = 0.50 + 0.20 = 0.70$.

### 2. The $\sigma$-algebra generated by $X$, namely $\sigma(X)$

The random variable $X$ records whether an item is good (0) or defective (1):
$X(G_1)=0, X(G_2)=0, X(D_1)=1, X(D_2)=1$.

To find $\sigma(X)$, we look at the preimages of the values $X$ can take:

*   $X=0$ corresponds to the set $X^{-1}(\{0\}) = \{G_1, G_2\}$ (the event that the item is Good).
*   $X=1$ corresponds to the set $X^{-1}(\{1\}) = \{D_1, D_2\}$ (the event that the item is Defective).

The $\sigma$-algebra generated by $X$, $\sigma(X)$, consists of $\emptyset$, $\Omega$, and all possible unions and complements of these basic sets. Thus:

$\sigma(X) = \{\emptyset, \{G_1, G_2\}, \{D_1, D_2\}, \Omega\}$.

### 3. The $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$

The random variable $Y$ records both quality and machine, assigning a unique value to each outcome:
$Y(G_1)=1, Y(G_2)=2, Y(D_1)=3, Y(D_2)=4$.

To find $\sigma(Y)$, we look at the preimages of the values $Y$ can take:

*   $Y=1$ corresponds to $Y^{-1}(\{1\}) = \{G_1\}$
*   $Y=2$ corresponds to $Y^{-1}(\{2\}) = \{G_2\}$
*   $Y=3$ corresponds to $Y^{-1}(\{3\}) = \{D_1\}$
*   $Y=4$ corresponds to $Y^{-1}(\{4\}) = \{D_2\}$

Since $Y$ distinguishes every individual element in the sample space, the sets that generate $\sigma(Y)$ are all the singletons. The $\sigma$-algebra generated by these singletons is the full power set:

$\sigma(Y) = \mathfrak{P}(\Omega)$ (all 16 subsets of $\Omega$).

### 5. Marginal Distribution of $X$

To find the marginal distribution of $X$, we sum the probabilities of the outcomes that correspond to each value of $X$.

- **$P(X=0)$**: This corresponds to the event $\{G_1, G_2\}$ (Good quality).  
  $P(X=0) = P(\{G_1\}) + P(\{G_2\}) = 0.50 + 0.20 = \mathbf{0.70}$

- **$P(X=1)$**: This corresponds to the event $\{D_1, D_2\}$ (Defective quality).  
  $P(X=1) = P(\{D_1\}) + P(\{D_2\}) = 0.10 + 0.20 = \mathbf{0.30}$

**Marginal Distribution of $X$:**  
$P(X=0) = 0.70$  
$P(X=1) = 0.30$

### 6. Marginal Distribution of $Y$

To find the marginal distribution of $Y$, we use the probabilities of the individual outcomes in $\Omega$, as $Y$ assigns a unique value to each.

- **$P(Y=1)$**: Corresponds to $\{G_1\}$.  
  $P(Y=1) = P(\{G_1\}) = \mathbf{0.50}$

- **$P(Y=2)$**: Corresponds to $\{G_2\}$.  
  $P(Y=2) = P(\{G_2\}) = \mathbf{0.20}$

- **$P(Y=3)$**: Corresponds to $\{D_1\}$.  
  $P(Y=3) = P(\{D_1\}) = \mathbf{0.10}$

- **$P(Y=4)$**: Corresponds to $\{D_2\}$.  
  $P(Y=4) = P(\{D_2\}) = \mathbf{0.20}$

**Marginal Distribution of $Y$:**  
$P(Y=1) = 0.50$  
$P(Y=2) = 0.20$  
$P(Y=3) = 0.10$  
$P(Y=4) = 0.20$

### 7. Joint Probabilities

We need to compute $P(X=x, Y=y)$. This is the probability of the intersection of the event $X=x$ and the event $Y=y$. Since $Y$ uniquely identifies each outcome, $P(X=x, Y=y)$ will be $P(\{ \omega \in \Omega : X(\omega)=x \text{ and } Y(\omega)=y \})$.

*   **$P(X=0, Y=1)$**:
    $Y=1$ corresponds to $\{G_1\}$. For $G_1$, $X=0$. So, $P(X=0, Y=1) = P(\{G_1\}) = \mathbf{0.50}$

*   **$P(X=0, Y=2)$**:
    $Y=2$ corresponds to $\{G_2\}$. For $G_2$, $X=0$. So, $P(X=0, Y=2) = P(\{G_2\}) = \mathbf{0.20}$

*   **$P(X=1, Y=3)$**:
    $Y=3$ corresponds to $\{D_1\}$. For $D_1$, $X=1$. So, $P(X=1, Y=3) = P(\{D_1\}) = \mathbf{0.10}$

*   **$P(X=1, Y=4)$**:
    $Y=4$ corresponds to $\{D_2\}$. For $D_2$, $X=1$. So, $P(X=1, Y=4) = P(\{D_2\}) = \mathbf{0.20}$

### 8. Conditional Probabilities

We use the formula: $P(A|B) = \frac{P(A \cap B)}{P(B)}$.

#### (a) $P(X=1 \mid Y=3)$
$P(X=1 \mid Y=3) = \frac{P(X=1, Y=3)}{P(Y=3)} = \frac{0.10}{0.10} = \mathbf{1}$

**Interpretation**: If we know the product is from **Machine 1 and Defective** (Y=3), it is **certain** to be Defective (X=1).

#### (b) $P(X=1 \mid Y=4)$
$P(X=1 \mid Y=4) = \frac{P(X=1, Y=4)}{P(Y=4)} = \frac{0.20}{0.20} = \mathbf{1}$

**Interpretation**: If we know the product is from **Machine 2 and Defective** (Y=4), it is **certain** to be Defective (X=1).

#### (c) $P(Y=3 \mid X=1)$
$P(Y=3 \mid X=1) = \frac{P(X=1, Y=3)}{P(X=1)} = \frac{0.10}{0.30} = \frac{1}{3} \approx 0.333$

**Interpretation**: If we know the product is **Defective** (X=1), there is a **1/3 chance** it came from **Machine 1** (Y=3).

#### (d) $P(Y=4 \mid X=1)$
$P(Y=4 \mid X=1) = \frac{P(X=1, Y=4)}{P(X=1)} = \frac{0.20}{0.30} = \frac{2}{3} \approx 0.667$

**Interpretation**: If we know the product is **Defective** (X=1), there is a **2/3 chance** it came from **Machine 2** (Y=4).

### 9. Interpretation of Conditioning

#### Conditioning on $Y$
When we condition on $Y$ (e.g., $P(X=1 \mid Y=3)$), we are given **very specific information** — the exact outcome in $\Omega$ (e.g., $D_1$).

Because $Y$ identifies each unique outcome, knowing the value of $Y$ **completely determines** the value of $X$. For example, if we know $Y=3$, we know it is $D_1$, which **must** have $X=1$.

Therefore, conditioning on $Y$ provides **"perfect" information** about $X$, eliminating all uncertainty about the quality.

#### Conditioning on $X$
When we condition on $X$ (e.g., $P(Y=3 \mid X=1)$), we are given information about the **quality** (e.g., "defective") but **not** the specific machine.

This narrows down the possibilities in $\Omega$ (from all four outcomes to only $\{D_1, D_2\}$), but it does **not eliminate all uncertainty** about $Y$. We still don’t know *which* machine produced the defective item.

The probabilities $P(Y=3 \mid X=1) = \frac{1}{3}$ and $P(Y=4 \mid X=1) = \frac{2}{3}$ reflect this **remaining uncertainty**.

### 10. $\sigma(X)$ vs $\sigma(Y)$ Information

#### Information available in $\sigma(X)$ but not in $\sigma(Y)$: **None**

As shown earlier, $\sigma(X) \subseteq \sigma(Y)$. This means that **any event** (piece of information) that can be described by $X$ can also be described by $Y$.

If $X$ knows something, $Y$ also knows it. For example:
- $X$ knows whether an item is "Good" or "Defective".
- $Y$ can also identify these categories by grouping its values ($Y=1,2$ for "Good"; $Y=3,4$ for "Defective").

#### Information available in $\sigma(Y)$ but not in $\sigma(X)$: **The specific Machine ID**

$\sigma(Y)$ contains **finer-grained information** than $\sigma(X)$.

While $\sigma(X)$ can distinguish between "Good" items $\{G_1, G_2\}$ and "Defective" items $\{D_1, D_2\}$, it **cannot differentiate within** those categories. For example, $\sigma(X)$ treats $G_1$ and $G_2$ as indistinguishable because they both fall under "Good".

In contrast, $\sigma(Y)$ can distinguish between individual outcomes like $G_1$ and $G_2$. This means $\sigma(Y)$ provides information about **which specific machine** (Machine 1 or Machine 2) produced the item, even if the quality is the same.

**Summary**:  
$Y$ knows both the **quality and the machine**, while $X$ only knows the **quality**.

### 12. Marginal, Joint, and Conditional Shannon Entropy of $X$ and $Y$

First, let's define the entropy function and then calculate the required values.

```python
import numpy as np
from scipy.stats import entropy

# Define the probabilities
p_G1 = 0.5
p_G2 = 0.2
p_D1 = 0.1
p_D2 = 0.2

# 1. Marginal Distribution of X
p_X0 = p_G1 + p_G2  # P(X=0) = P(Good)
p_X1 = p_D1 + p_D2  # P(X=1) = P(Defective)

# 2. Marginal Distribution of Y
p_Y1 = p_G1
p_Y2 = p_G2
p_Y3 = p_D1
p_Y4 = p_D2

# --- Marginal Entropies ---
# Entropy of X
H_X = entropy([p_X0, p_X1], base=2)
print(f"H(X) = {H_X:.3f} bits")

# Entropy of Y
H_Y = entropy([p_Y1, p_Y2, p_Y3, p_Y4], base=2)
print(f"H(Y) = {H_Y:.3f} bits")

# --- Joint Probability Distribution P(X,Y) ---
# This is already effectively our original P(omega) since X and Y are functions on omega
# P(X=0, Y=1) = p_G1 = 0.5
# P(X=0, Y=2) = p_G2 = 0.2
# P(X=1, Y=3) = p_D1 = 0.1
# P(X=1, Y=4) = p_D2 = 0.2
# All other joint probabilities P(X=x, Y=y) where x is inconsistent with y are 0.

p_XY = [
    p_G1,  # P(X=0, Y=1)
    p_G2,  # P(X=0, Y=2)
    p_D1,  # P(X=1, Y=3)
    p_D2   # P(X=1, Y=4)
]

# --- Joint Entropy H(X,Y) ---
# The joint entropy H(X,Y) is the entropy of the joint distribution P(X,Y).
# Since (X,Y) effectively maps to the original outcomes, H(X,Y) is just the entropy of the original P(omega).
H_XY = entropy(p_XY, base=2)
print(f"H(X,Y) = {H_XY:.3f} bits")

# --- Conditional Entropies ---
# H(X|Y) = H(X,Y) - H(Y)
H_X_given_Y = H_XY - H_Y
print(f"H(X|Y) = H(X,Y) - H(Y) = {H_X_given_Y:.3f} bits")

# H(Y|X) = H(X,Y) - H(X)
H_Y_given_X = H_XY - H_X
print(f"H(Y|X) = H(X,Y) - H(X) = {H_Y_given_X:.3f} bits")

print("\n--- Detailed Conditional Probabilities for Entropy ---")

# Conditional Entropy H(X|Y)
# H(X|Y) = sum_y P(Y=y) * H(X|Y=y)
# We know X is completely determined by Y, so H(X|Y=y) should be 0 for all y.

# H(X|Y=1) = entropy([P(X=0|Y=1), P(X=1|Y=1)]) = entropy([1, 0]) = 0
# H(X|Y=2) = entropy([P(X=0|Y=2), P(X=1|Y=2)]) = entropy([1, 0]) = 0
# H(X|Y=3) = entropy([P(X=0|Y=3), P(X=1|Y=3)]) = entropy([0, 1]) = 0
# H(X|Y=4) = entropy([P(X=0|Y=4), P(X=1|Y=4)]) = entropy([0, 1]) = 0

# So H(X|Y) = 0 * p_Y1 + 0 * p_Y2 + 0 * p_Y3 + 0 * p_Y4 = 0.
# This matches H_X_given_Y calculated above, confirming Y gives full information about X.

# Conditional Entropy H(Y|X)
# H(Y|X) = sum_x P(X=x) * H(Y|X=x)

# H(Y|X=0) = entropy([P(Y=1|X=0), P(Y=2|X=0)])
# P(Y=1|X=0) = P(X=0,Y=1) / P(X=0) = p_G1 / p_X0 = 0.5 / 0.7 = 5/7
# P(Y=2|X=0) = P(X=0,Y=2) / P(X=0) = p_G2 / p_X0 = 0.2 / 0.7 = 2/7
H_Y_given_X0 = entropy([p_G1/p_X0, p_G2/p_X0], base=2)
print(f"H(Y|X=0) = {H_Y_given_X0:.3f} bits")

# H(Y|X=1) = entropy([P(Y=3|X=1), P(Y=4|X=1)])
# P(Y=3|X=1) = P(X=1,Y=3) / P(X=1) = p_D1 / p_X1 = 0.1 / 0.3 = 1/3
# P(Y=4|X=1) = P(X=1,Y=4) / P(X=1) = p_D2 / p_X1 = 0.2 / p_X1 = 2/3
H_Y_given_X1 = entropy([p_D1/p_X1, p_D2/p_X1], base=2)
print(f"H(Y|X=1) = {H_Y_given_X1:.3f} bits")

# Weighted sum for H(Y|X)
calculated_H_Y_given_X = p_X0 * H_Y_given_X0 + p_X1 * H_Y_given_X1
print(f"Calculated H(Y|X) = {calculated_H_Y_given_X:.3f} bits (should match H_Y_given_X above)")

# Mutual Information I(X;Y)
I_XY = H_X + H_Y - H_XY
print(f"I(X;Y) = {I_XY:.3f} bits")

# Alternatively, I(X;Y) = H(X) - H(X|Y)
alt_I_XY_X_H_X_H_X_given_Y = H_X - H_X_given_Y
print(f"I(X;Y) (H(X) - H(X|Y)) = {alt_I_XY_X_H_X_H_X_given_Y:.3f} bits")

# Alternatively, I(X;Y) = H(Y) - H(Y|X)
alt_I_XY_Y_H_Y_H_Y_given_X = H_Y - H_Y_given_X
print(f"I(X;Y) (H(Y) - H(Y|X)) = {alt_I_XY_Y_H_Y_H_Y_given_X:.3f} bits")
```
