## Bertrand's Paradox: Introduction

Bertrand's Paradox explains how a probability problem can have different solutions depending on how the terms of the problem are defined.
<br>
Specifically, it asks for the probability that a 'random' chord of a circle is longer than the side of an equilateral triangle inscribed in the same circle.
<br>
The ambiguity arises from the definition of 'random chord'.

## Geometrical Setup

Let's consider a circle with radius $R$.
<br>
 An equilateral triangle inscribed in this circle has a side length $L = R\sqrt{3}$.

A chord is longer than $L$ if its distance from the center of the circle is less than $R/2$.

To see this, consider an equilateral triangle inscribed in the circle. The height of this triangle from the center to the midpoint of a side is $R/2$. Any chord whose distance from the center is less than $R/2$ will be longer than the side of the inscribed equilateral triangle.

We will use a unit circle (radius $R=1$) for simplicity.
<br>
The side length of the inscribed equilateral triangle is $\sqrt{3}$.
<br>
A chord is longer than $\sqrt{3}$ if its distance from the center is less than $1/2$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Parameters for a unit circle
R = 1
side_equilateral = R * np.sqrt(3)

# Condition for a chord to be longer than the equilateral triangle side:
# Its distance from the center must be less than R/2
condition_distance = R / 2

print(f"Radius of the circle (R): {R}")
print(f"Side length of inscribed equilateral triangle: {side_equilateral:.4f}")
print(f"Distance from center for a chord of length {side_equilateral:.4f}: {condition_distance:.4f}")
```

## Method 1: Random Endpoints

Choose two random points on the circumference of the circle and connect them to form a chord.

To simulate this:
1.  Generate two random angles $\theta_1$ and $\theta_2$ (between $0$ and $2\pi$).
2.  These angles define two points $(R\cos\theta_1, R\sin\theta_1)$ and $(R\cos\theta_2, R\sin\theta_2)$.
3.  Calculate the length of the chord using the distance formula.
4.  Alternatively, the length of a chord connecting two points with angular separation $\Delta\theta$ is $2R\sin(\Delta\theta/2)$. The chord is longer than the equilateral triangle side if $2R\sin(\Delta\theta/2) > R\sqrt{3}$, which simplifies to $\sin(\Delta\theta/2) > \sqrt{3}/2$. This means $\Delta\theta/2$ must be between $\pi/3$ and $2\pi/3$. So, $\Delta\theta$ must be between $2\pi/3$ and $4\pi/3$. The total range for $\Delta\theta$ is $2\pi$. Thus, the probability is $(4\pi/3 - 2\pi/3) / (2\pi) = (2\pi/3) / (2\pi) = 1/3$.

Let's simulate this method.

```python
# Method 1: Random Endpoints
num_simulations = 100000

longer_chords_m1 = 0
for _ in range(num_simulations):
    # Choose two random angles for the endpoints
    theta1 = np.random.uniform(0, 2 * np.pi)
    theta2 = np.random.uniform(0, 2 * np.pi)

    # Calculate the angular separation
    delta_theta = np.abs(theta1 - theta2)
    # The angular separation can be measured in two directions around the circle.
    # We take the smaller one for chord length calculation.
    delta_theta = min(delta_theta, 2 * np.pi - delta_theta)

    # Calculate chord length
    chord_length = 2 * R * np.sin(delta_theta / 2)

    if chord_length > side_equilateral:
        longer_chords_m1 += 1

probability_m1 = longer_chords_m1 / num_simulations
print(f"Probability (Method 1 - Random Endpoints): {probability_m1:.4f}")
```

## Method 2: Random Radius and Point

Choose a random radius of the circle, and then choose a random point on this radius. Construct the chord perpendicular to the chosen radius, passing through the chosen point.

To simulate this:
1.  Choose a random radius by picking an angle $\theta$ between $0$ and $2\pi$.
2.  Choose a random point along this radius by picking a distance $d$ from the center between $0$ and $R$.
3.  The chord's distance from the center is $d$.
4.  The chord is longer than the equilateral triangle side if $d < R/2$.

In this method, the probability is simply the ratio of the favorable distance range to the total distance range: $(R/2) / R = 1/2$.

```python
# Method 2: Random Radius and Point
longer_chords_m2 = 0
for _ in range(num_simulations):
    # Choose a random distance from the center (d) along a radius
    # The angle of the radius is irrelevant for the distance of the chord.
    d = np.random.uniform(0, R)

    if d < condition_distance:
        longer_chords_m2 += 1

probability_m2 = longer_chords_m2 / num_simulations
print(f"Probability (Method 2 - Random Radius and Point): {probability_m2:.4f}")
```

## Method 3: Random Midpoint

Choose a random point anywhere within the circle. This point will be the midpoint of the chord. Construct the chord perpendicular to the radius passing through this midpoint.

To simulate this:
1.  Choose a random point $(x, y)$ within the circle. This can be done by picking a random radius $r$ between $0$ and $R$ and a random angle $\theta$ between $0$ and $2\pi$, so $(x,y) = (r\cos\theta, r\sin\theta)$.
2.  The distance of the chord from the center is $r$.
3.  The chord is longer than the equilateral triangle side if $r < R/2$.

In this method, the probability is the ratio of the area of the concentric circle with radius $R/2$ to the area of the main circle with radius $R$. The ratio of the areas is $( \pi (R/2)^2 ) / ( \pi R^2 ) = (R^2/4) / R^2 = 1/4$.

```python
# Method 3: Random Midpoint
longer_chords_m3 = 0
for _ in range(num_simulations):
    # Choose a random point (x, y) within the circle
    # Easier to do by choosing a random radius r and angle theta
    r_rand = np.random.uniform(0, R) # Distance from center
    theta_rand = np.random.uniform(0, 2 * np.pi)

    # The distance of the chord from the center is r_rand
    d = r_rand

    if d < condition_distance:
        longer_chords_m3 += 1

probability_m3 = longer_chords_m3 / num_simulations
print(f"Probability (Method 3 - Random Midpoint): {probability_m3:.4f}")
```

## Summary of Results

As demonstrated by the simulations:
*   **Method 1 (Random Endpoints):** Probability $\approx 1/3$
*   **Method 2 (Random Radius and Point):** Probability $\approx 1/2$
*   **Method 3 (Random Midpoint):** Probability $\approx 1/4$

Each of these methods provides a 'natural' way to define a random chord, yet they yield different probabilities. This is the essence of Bertrand's Paradox, highlighting the importance of precisely defining the sample space and the probability measure when dealing with geometric probabilities.
