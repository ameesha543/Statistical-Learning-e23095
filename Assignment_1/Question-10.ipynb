### 1. Probability Space $(\Omega, \mathfrak{F}, P)$ Explicitly

The problem explicitly defines the probability space as follows:

- **Sample Space**: $\Omega = \mathbb{R}^2$, representing all possible pairs of temperature and pressure deviations $(t, p)$.

- **$\sigma$-algebra**: $\mathfrak{F} = \mathcal{B}(\mathbb{R}^2)$, the Borel $\sigma$-algebra on $\mathbb{R}^2$. This consists of all measurable subsets of $\mathbb{R}^2$ for which we can assign a probability.

- **Probability Measure $P$**: The probability measure induced by the joint probability density function $f_{T,P}(t,p)$.  
  For any measurable set $A \in \mathcal{B}(\mathbb{R}^2)$, the probability is given by:
  
  $$
  P(A) = \iint_{A} f_{T,P}(t,p) \, dt \, dp = \iint_{A} \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right) \, dt \, dp
  $$

### 2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$

The random variable $X:\Omega\to\mathbb R$ is defined by $X(t,p)=t$. The $\sigma$-algebra generated by $X$, denoted $\sigma(X)$, is the collection of all preimages of Borel sets in $\mathbb R$ under $X$. That is:

$$
\sigma(X) = \{ X^{-1}(B) \mid B \in \mathfrak B(\mathbb R) \}
$$

For any Borel set $B \subseteq \mathbb R$:

$$
X^{-1}(B) = \{ (t,p) \in \mathbb R^2 \mid X(t,p) \in B \} = \{ (t,p) \in \mathbb R^2 \mid t \in B \} = B \times \mathbb R
$$

Thus, $\sigma(X)$ consists of all infinite strips in $\mathbb R^2$ where the first coordinate $t$ belongs to some Borel set $B$. Formally:

$$
\sigma(X) = \{ B \times \mathbb R \mid B \in \mathfrak B(\mathbb R) \}
$$


### 3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$

The random variable $Y:\Omega\to\mathbb R^2$ is defined by $Y(t,p)=(t,p)$. The $\sigma$-algebra generated by $Y$, denoted $\sigma(Y)$, is the collection of all preimages of Borel sets in $\mathbb R^2$ under $Y$. That is:

$$
\sigma(Y) = \{ Y^{-1}(C) \mid C \in \mathfrak B(\mathbb R^2) \}
$$

For any Borel set $C \subseteq \mathbb R^2$:

$$
Y^{-1}(C) = \{ (t,p) \in \mathbb R^2 \mid Y(t,p) \in C \} = \{ (t,p) \in \mathbb R^2 \mid (t,p) \in C \} = C
$$

Thus, $\sigma(Y)$ is simply the Borel $\sigma$-algebra on $\mathbb R^2$ itself:

$$
\sigma(Y) = \mathfrak B(\mathbb R^2)
$$


### 4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$

**Showing $\sigma(X) \subseteq \sigma(Y)$:**

From **Question 2**, we have $\sigma(X) = \{ B \times \mathbb R \mid B \in \mathfrak B(\mathbb R) \}$.
From **Question 3**, we have $\sigma(Y) = \mathfrak B(\mathbb R^2)$.

Any set of the form $B \times \mathbb R$, where $B$ is a Borel set in $\mathbb R$, is itself a Borel set in $\mathbb R^2$. This is because the product of Borel sets is a Borel set. Therefore, every set in $\sigma(X)$ is also a set in $\mathfrak B(\mathbb R^2) = \sigma(Y)$.

Thus, we have shown that $\sigma(X)\subseteq \sigma(Y)$.

**Explanation of Information Content:**

The $\sigma$-algebra generated by a random variable represents the

volume of information available from that random variable. A coarser $\sigma$-algebra implies less information, as fewer distinctions can be made in the sample space.

Since $\sigma(X) \subseteq \sigma(Y)$, it means that all events (measurable sets) that can be defined using $X$ can also be defined using $Y$. However, $Y$ can define more events (all of $\mathfrak B(\mathbb R^2)$) than $X$ (only strips $B \times \mathbb R$). This implies that $Y$ allows for a finer partitioning of the sample space, hence providing more detailed information.

**Analogy:**
*   Knowing $X$ (temperature only) is like knowing which *row* you are in on a grid, but not which *column*. You can distinguish between different rows but not between points within the same row.
*   Knowing $Y$ (temperature and pressure) is like knowing the exact *cell* on the grid. You can pinpoint the exact location in $\mathbb R^2$.

Thus, $Y$ contains strictly more information than $X$ because it captures both dimensions of the sensor state, whereas $X$ only captures one.

### 8. Compute the conditional distribution of $X$ given $Y=(t,p)$

If we know that $Y=(t,p)$ exactly, then we know both the temperature deviation $t$ and the pressure deviation $p$. Since $X(t,p)=t$, knowing $Y=(t,p)$ implies that $X$ must be exactly $t$. The conditional distribution is therefore a deterministic value.

In the context of probability distributions, this is represented by a **Dirac delta distribution** centered at $t$: $\delta(x - t)$.

This means that $P(X \in A \mid Y=(t,p)) = 1$ if $t \in A$, and $0$ otherwise.

### 9. Compute the conditional distribution of $Y$ given $X=t$

If we know that $X=t$, it means the temperature deviation is fixed at $t$. However, the pressure deviation $P$ is still a random variable. Since $T$ and $P$ are independent, knowing $T=t$ gives us no information about $P$. Therefore, $P$ retains its marginal distribution $N(0,4)$.

The conditional distribution of $Y=(T,P)$ given $X=t$ (which is $T=t$) is a random vector where the first component is fixed at $t$ and the second component is a random variable following $N(0,4)$.

$$Y \mid (X=t) \sim (t, N(0,4))$$

This can be thought of as a 1D normal distribution along the vertical line $T=t$ in the $(t,p)$-plane, with density $f_P(p)$ scaled by the Dirac delta $\delta(t-t')$.

Formally, the conditional PDF $f_{Y|X}(t',p'|t)$ would be:

$$f_{Y|X}(t',p'|t) = \delta(t'-t) \cdot f_P(p') = \delta(t'-t) \cdot \frac{1}{\sqrt{2\pi \cdot 4}}\exp\left(-\frac{p'^2}{2 \cdot 4}\right)$$

### 11. Show that $X=\pi_1\circ Y$, where $\pi_1(t,p)=t$, and explain why this implies that $X$ is a measurable function of $Y$.

Let's verify the composition:

We have $Y:\Omega \to \mathbb R^2$ defined by $Y(t',p') = (t',p')$.
And $\pi_1:\mathbb R^2 \to \mathbb R$ defined by $\pi_1(t,p) = t$.

Then, the composition $X_{comp} = \pi_1 \circ Y$ is a mapping from $\Omega$ to $\mathbb R$:

$$(\pi_1 \circ Y)(t',p') = \pi_1(Y(t',p')) = \pi_1(t',p') = t'$$

This is exactly the definition of our random variable $X$: $X(t',p') = t'$.

Thus, we have shown that $X = \pi_1 \circ Y$.

**Why this implies $X$ is a measurable function of $Y$:**

In probability theory, a random variable $X$ is said to be a measurable function of another random variable $Y$ if $X$ can be expressed as $X = g(Y)$ for some measurable function $g$. In our case, $g = \pi_1$.

The projection function $\pi_1(t,p)=t$ is a continuous function. A fundamental property in measure theory is that all continuous functions from one measurable space to another are Borel-measurable functions. Since $\pi_1$ is continuous, it is measurable with respect to the Borel $\sigma$-algebras $\mathfrak B(\mathbb R^2)$ and $\mathfrak B(\mathbb R)$.

When $X = g(Y)$ where $g$ is measurable, it implies that $\sigma(X) \subseteq \sigma(Y)$. This means that any information revealed by $X$ is already contained within the information revealed by $Y$. In simpler terms, if you know the value of $Y$, you can unambiguously determine the value of $X$ by applying the function $g$ (in this case, taking the first component). Therefore, $X$ does not introduce any new randomness or information not already present in $Y$.

### 13. Compute the differential entropy of $X$

For a normal distribution $N(\mu, \sigma^2)$, the differential entropy in *nats* (natural units of information, since we use the natural logarithm) is given by the formula:

$$h(Z) = \frac{1}{2}\ln(2\pi e \sigma^2)$$

Here, $X \sim N(0,1)$, so $\mu = 0$ and $\sigma^2 = 1$.

$$h(X) = \frac{1}{2}\ln(2\pi e \cdot 1) = \frac{1}{2}\ln(2\pi e)$$

Using numerical values:
$$h(X) \approx \frac{1}{2}\ln(2 \cdot 3.14159 \cdot 2.71828) \approx \frac{1}{2}\ln(17.079) \approx \frac{1}{2} \cdot 2.8377 \approx \mathbf{1.4189 \text{ nats}}$$


### 14. Compute the differential entropy of $Y$

$Y=(T,P)$ is a 2-dimensional random vector. Since $T \sim N(0,1)$ and $P \sim N(0,4)$ are independent, $Y$ follows a multivariate normal distribution with mean vector $\mu = (0,0)^T$ and a diagonal covariance matrix $\Sigma = \begin{pmatrix} 1 & 0 \\ 0 & 4 \end{pmatrix}$.

The determinant of the covariance matrix is $|\Sigma| = 1 \cdot 4 - 0 \cdot 0 = 4$.

For a $k$-dimensional multivariate normal distribution, the differential entropy is:

$$h(Y) = \frac{1}{2}\ln((2\pi e)^k |\Sigma|)$$

Here, $k=2$.

$$h(Y) = \frac{1}{2}\ln((2\pi e)^2 \cdot 4) = \frac{1}{2}\ln(4 \cdot (2\pi e)^2)$$
$$h(Y) = \frac{1}{2}\ln( (2 \cdot 2\pi e)^2 ) = \frac{1}{2} \cdot 2 \ln(4\pi e) = \ln(4\pi e)$$

Alternatively, since $T$ and $P$ are independent, the joint entropy is the sum of the individual entropies:
$$h(Y) = h(T) + h(P)$$
We already found $h(T) = h(X) = \frac{1}{2}\ln(2\pi e)$.
For $P \sim N(0,4)$, $\sigma^2 = 4$.
$$h(P) = \frac{1}{2}\ln(2\pi e \cdot 4) = \frac{1}{2}\ln(8\pi e)$$
So,
$$h(Y) = \frac{1}{2}\ln(2\pi e) + \frac{1}{2}\ln(8\pi e) = \frac{1}{2} (\ln(2\pi e) + \ln(8\pi e)) = \frac{1}{2} \ln((2\pi e)(8\pi e))$$
$$h(Y) = \frac{1}{2} \ln(16\pi^2 e^2) = \frac{1}{2} \ln((4\pi e)^2) = \ln(4\pi e)$$

Using numerical values:
$$h(Y) \approx \ln(4 \cdot 3.14159 \cdot 2.71828) \approx \ln(34.159) \approx \mathbf{3.535 \text{ nats}}$$


### 15. Compute the conditional differential entropy $h(Y|X)$, and interpret it physically.

The conditional differential entropy $h(Y|X)$ measures the remaining uncertainty in $Y$ once $X$ is known. In our case, $Y=(T,P)$ and $X=T$. So, $h(Y|X)$ is the uncertainty in $(T,P)$ given $T$.

Using the chain rule for differential entropy, we have:
$$h(Y|X) = h(T,P|T)$$
Since $T$ is given, its uncertainty is zero. The remaining uncertainty is only about $P$.

More rigorously, by the definition of conditional entropy for jointly continuous variables:
$$h(Y|X) = \int_{\mathbb R} f_X(t) \left( -\int_{\mathbb R^2} f_{Y|X}(t',p'|t) \ln f_{Y|X}(t',p'|t) \,dt'\,dp' \right) \,dt$$

From Question 9, we found that $f_{Y|X}(t',p'|t) = \delta(t'-t) \cdot f_P(p')$.
Substituting this into the inner integral:
$$-\int_{\mathbb R^2} \delta(t'-t) f_P(p') \ln(\delta(t'-t) f_P(p')) \,dt'\,dp'$$
This integral is problematic due to the Dirac delta function. A more practical approach for independent components is to recognize that $Y=(X,P)$.

So, $h(Y|X) = h(X,P|X)$. When $X$ is known, $X$ is no longer a random variable, and the uncertainty in the pair $(X,P)$ reduces to the uncertainty in $P$.

Therefore, $h(Y|X) = h(P|X)$. Since $X$ (which is $T$) and $P$ are independent, knowing $X$ provides no information about $P$. Thus, $h(P|X) = h(P)$.

We know $P \sim N(0,4)$, so its variance is $\sigma_P^2 = 4$. The differential entropy of $P$ is:
$$h(P) = \frac{1}{2}\ln(2\pi e \sigma_P^2) = \frac{1}{2}\ln(2\pi e \cdot 4) = \frac{1}{2}\ln(8\pi e)$$

Using numerical values:
$$h(P) \approx \frac{1}{2}\ln(8 \cdot 3.14159 \cdot 2.71828) \approx \frac{1}{2}\ln(68.21) \approx \frac{1}{2} \cdot 4.222 \approx \mathbf{2.111 \text{ nats}}$$

**Physical Interpretation:**

When we condition on $X$ (the temperature deviation), we fix its value. The only remaining uncertainty in the overall system state $Y=(T,P)$ is due to the pressure deviation $P$. Since $T$ and $P$ are independent, knowing $T$ doesn't reduce the uncertainty about $P$ at all. Therefore, $h(Y|X)$ is exactly the differential entropy of the pressure deviation $P$. It quantifies the amount of

randomness

remaining in the pressure measurement even after the temperature measurement is known.

### 16. Discuss what happens to $h(X|Y)$ in this model, and explain why the continuous case differs from the discrete case when one random variable is completely determined by the other.

In this model, $X(t,p) = t$ and $Y(t,p) = (t,p)$. This means that if we know $Y$, we *perfectly* know $X$. That is, $X$ is a deterministic function of $Y$. If $Y=(t_0, p_0)$, then $X$ must be $t_0$.

**Discrete Case Analogy:**
If we had discrete random variables, say $X_D$ and $Y_D$, and $X_D$ was completely determined by $Y_D$, then the conditional entropy $H(X_D|Y_D)$ would be 0 bits. This is because there is no uncertainty left in $X_D$ once $Y_D$ is known. All outcomes for $X_D$ are perfectly predictable.

**Continuous Case ($h(X|Y)$):**
In the continuous case, the situation is different and leads to a concept that can be counter-intuitive. If $X$ is perfectly determined by $Y$, the conditional distribution of $X$ given $Y=(t,p)$ is a **Dirac delta function** centered at $t$. This is a distribution where all its probability mass is concentrated at a single point.

The differential entropy of a distribution, $h(Z) = -\int f_Z(z) \ln f_Z(z) \,dz$, measures the "spread" or "volume" of the probability density. For a Dirac delta function, which is infinitely narrow and infinitely high at a single point, the concept of a finite "spread" breaks down. As the variance of a continuous distribution approaches zero (making it more and more concentrated at a single point), its differential entropy tends to **$-\infty$**.

So, in this model, $h(X|Y) = -\infty$.

**Why the continuous case differs from the discrete case:**

1.  **Nature of "Uncertainty":**
    *   **Discrete Entropy ($H$):** Measures the *average number of bits* needed to describe the outcome of a random variable. When an outcome is perfectly known, 0 bits are needed. It's a measure of absolute uncertainty.
    *   **Differential Entropy ($h$):** Measures the *relative spread or volume* of a probability density function. It's not an absolute measure of uncertainty like its discrete counterpart. It depends on the coordinate system and can be negative. A value of $-\infty$ indicates that the probability mass is concentrated on a set of measure zero (a single point), meaning there's no "volume" of uncertainty.

2.  **Point Probability:**
    *   In discrete probability, a single point can have a non-zero probability. Knowing that point means the uncertainty is zero.
    *   In continuous probability, the probability of any single point is zero. A deterministic continuous variable corresponds to a probability density that is zero everywhere except at one point, where it's infinitely high (a Dirac delta). This mathematical object, while useful, causes the entropy integral to diverge to $-\infty$.

3.  **Information Content:**
    When $X$ is a deterministic function of $Y$, there is no *additional* information gained about $X$ from knowing $Y$; $X$ is already known. The $-\infty$ value for $h(X|Y)$ in the continuous case can be interpreted as having infinite precision or absolute certainty, which is a mathematical idealization not achievable with finite-precision measurements. This divergence highlights a key conceptual difference between discrete and continuous information theory when dealing with variables that are perfectly determined by others.
