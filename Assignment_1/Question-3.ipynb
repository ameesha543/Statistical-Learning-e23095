# Transformer Failures Analysis

 1. Probability Space $(\Omega, \mathfrak{F}, P)$<br> $X \in \{0, 1, 2, 3\}$
- $\Omega$: The set of all possible outcomes for the number of failures: $\Omega = \{0, 1, 2, 3\}$.
- $\mathfrak{F}$: The power set of $\Omega$, $\mathcal{P}(\Omega)$, which contains all $2^4 = 16$ possible subsets of outcomes.
- $P$: A function $P: \mathfrak{F} \to [0, 1]$ defined by the given probabilities for each elementary event: $P(\{0\})=0.50, P(\{1\})=0.30, P(\{2\})=0.15, P(\{3\})=0.05$.

### 2. Definition of $X$
$X$ is defined as a random variable mapping from the sample space to the real numbers<br> $X: \Omega \to \mathbb{R}$,

### 3. Probability Mass Function (PMF)
The PMF $f_X(x) = P(X=x)$ for this distribution is:
- $f_X(0) = 0.50$
- $f_X(1) = 0.30$
- $f_X(2) = 0.15$
- $f_X(3) = 0.05$


```python
import pandas as pd
import numpy as np

# Define the distribution
x_values = np.array([0, 1, 2, 3])
probs = np.array([0.50, 0.30, 0.15, 0.05])

# 3. PMF
pmf = pd.Series(probs, index=x_values)
print("Probability Mass Function (PMF):")
print(pmf)

# 4. E[X]
expectation = np.sum(x_values * probs)

# 5. Var(X)
variance = np.sum(((x_values - expectation)**2) * probs)

# 6. P(X >= 1) and P(X >= 2)
p_ge_1 = np.sum(probs[x_values >= 1])
p_ge_2 = np.sum(probs[x_values >= 2])

print(f"\nExpectation E[X]: {expectation:.2f}")
print(f"Variance Var(X): {variance:.4f}")
print(f"P(X >= 1): {p_ge_1:.2f}")
print(f"P(X >= 2): {p_ge_2:.2f}")
```

### 7. Practical Meaning

*   **Expectation:** This represents the long-term average number of emergency transformer failures per day. If the district records failures over many years, the average would settle around 0.75 failures/day.
*   **Variance:** This measures the spread or uncertainty. A higher variance would indicate that the number of daily failures fluctuates significantly from the average, while a lower variance would mean the number of failures is more consistent.
