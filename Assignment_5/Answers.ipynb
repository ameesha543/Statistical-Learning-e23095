# 1. Bias of the Maximum Likelihood Estimator (MLE) of the Covariance Matrix

Let

$$
\widehat{\boldsymbol{\mu}}_n
=
\frac{1}{n}
\sum_{i=1}^{n}
\mathbf{X}_i
$$

and define the Maximum Likelihood Estimator (MLE) of the covariance matrix as

$$
\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}
=
\frac{1}{n}
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T.
$$

To determine whether this estimator is unbiased, consider the identity

$$
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
=
\sum_{i=1}^{n}
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
-
n
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T.
$$

Taking expectations on both sides,

$$
\mathbb{E}
\left[
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
\right]
=
\mathbb{E}
\left[
\sum_{i=1}^{n}
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
\right]
-
n
\mathbb{E}
\left[
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
\right].
$$

Since

$$
\mathbb{E}
\left[
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
\right]
=
\boldsymbol{\Sigma},
$$

we obtain

$$
\mathbb{E}
\left[
\sum_{i=1}^{n}
(\mathbf{X}_i-\boldsymbol{\mu})
(\mathbf{X}_i-\boldsymbol{\mu})^T
\right]
=
n\boldsymbol{\Sigma}.
$$

Also,

$$
\mathrm{Var}
(\widehat{\boldsymbol{\mu}}_n)
=
\frac{\boldsymbol{\Sigma}}{n},
$$

which implies

$$
\mathbb{E}
\left[
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})^T
\right]
=
\frac{\boldsymbol{\Sigma}}{n}.
$$

Substituting,

$$
\mathbb{E}
\left[
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
\right]
=
n\boldsymbol{\Sigma}
-
n
\left(
\frac{\boldsymbol{\Sigma}}{n}
\right)
=
(n-1)\boldsymbol{\Sigma}.
$$

Therefore,

$$
\mathbb{E}
[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{1}{n}
(n-1)\boldsymbol{\Sigma}
=
\frac{n-1}{n}
\boldsymbol{\Sigma}.
$$

Hence,

$$
\boxed{
\mathbb{E}
[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]
=
\frac{n-1}{n}
\boldsymbol{\Sigma}
}
$$

which proves that the raw MLE underestimates the true covariance matrix and is therefore biased.

---

## Bessel's Correction and the Unbiased Estimator

To eliminate this bias, we replace the denominator \(n\) with \(n-1\). The corrected estimator is

$$
\mathbf{S}
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T.
$$

Taking expectations,

$$
\mathbb{E}[\mathbf{S}]
=
\frac{1}{n-1}
\mathbb{E}
\left[
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T
\right].
$$

Using the result derived above,

$$
\mathbb{E}[\mathbf{S}]
=
\frac{1}{n-1}
(n-1)\boldsymbol{\Sigma}
=
\boldsymbol{\Sigma}.
$$

Therefore,

$$
\boxed{
\mathbb{E}[\mathbf{S}]
=
\boldsymbol{\Sigma}
}
$$

showing that \(\mathbf{S}\) is an unbiased estimator of the population covariance matrix.

The correction is called **Bessel's correction** and compensates for the loss of one degree of freedom caused by estimating the population mean from the same sample.

# 2. Type I and Type II Errors in Structural Health Monitoring

Assume the null hypothesis is

$$
H_0:\text{The structure is healthy}.
$$

and the alternative hypothesis is

$$
H_1:\text{The structure is damaged}.
$$

## Type I Error (\(\alpha\))

A Type I Error occurs when the monitoring system rejects \(H_0\) even though the structure is actually healthy.

Mathematically,

$$
\boxed{
\alpha
=
P(\text{Reject } H_0 \mid H_0 \text{ is true})
}
$$

In structural health monitoring, this corresponds to a **false alarm**, where a healthy structure is incorrectly classified as damaged. Such errors may trigger unnecessary inspections, maintenance procedures, or costly plant shutdowns.

---

## Type II Error (\(\beta\))

A Type II Error occurs when the monitoring system fails to reject \(H_0\) even though damage is actually present.

Mathematically,

$$
\boxed{
\beta
=
P(\text{Fail to reject } H_0 \mid H_1 \text{ is true})
}
$$

In structural health monitoring, this corresponds to **missed damage detection**, where an unhealthy structure is incorrectly classified as healthy.

---

## Effect of an Ultra-Conservative Significance Level

Suppose the engineer chooses

$$
\alpha = 0.0001.
$$

The statistical power of the test is

$$
\boxed{
\text{Power}
=
1-\beta
}
$$

which measures the probability of correctly detecting actual damage.

Reducing \(\alpha\) makes the rejection region much smaller because stronger evidence is required before damage is declared.

Consequently,

$$
\alpha \downarrow
\quad \Longrightarrow \quad
\beta \uparrow
\quad \Longrightarrow \quad
(1-\beta) \downarrow.
$$

Thus, lowering the significance level decreases the probability of false alarms but also reduces the probability of detecting real damage.

---

## Effect on the Healthy-Operation Confidence Ellipsoid

In multivariate monitoring systems, healthy observations are often represented by a confidence ellipsoid defined by

$$
(\mathbf{x}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}
(\mathbf{x}-\boldsymbol{\mu})
\leq c_{\alpha},
$$

where \(c_{\alpha}\) is a critical threshold chosen so that

$$
P
\left(
(\mathbf{X}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}

\leq c_{\alpha}
\right)
=
1-\alpha.
$$

When \(\alpha\) is reduced,

$$
1-\alpha
\rightarrow 1,
$$

which increases the critical value \(c_{\alpha}\).

Therefore,

$$
\boxed{
\alpha \downarrow
\quad \Longrightarrow \quad
c_{\alpha} \uparrow
}
$$

and the healthy-operation ellipsoid expands.

Since the volume of the ellipsoid is proportional to

$$
V
\propto
(c_{\alpha})^{m/2},
$$

where \(m\) is the dimension of the feature space,

$$
\boxed{
V
\propto
(c_{\alpha})^{m/2}
}
$$

a larger critical value produces a larger confidence ellipsoid.

Therefore, choosing

$$
\alpha = 0.0001
$$

results in:

- A much larger healthy-operation region.
- Fewer false alarms.
- Lower statistical power.
- Greater risk of missing actual structural damage.

Thus, an ultra-conservative significance threshold improves operational stability but reduces damage-detection sensitivity.

# Slutsky's Theorem and the Practical Distribution of the Sample Mean

The Multivariate Central Limit Theorem states that for a sequence of i.i.d. random vectors

$$
\mathbf{X}_1,\mathbf{X}_2,\ldots,\mathbf{X}_n \in \mathbb{R}^m,
$$

with population mean

$$
\boldsymbol{\mu}
=
\mathbb{E}[\mathbf{X}_i]
$$

and covariance matrix

$$
\boldsymbol{\Sigma}
=
\mathrm{Var}(\mathbf{X}_i),
$$

the sample mean

$$
\widehat{\boldsymbol{\mu}}_n
=
\frac{1}{n}
\sum_{i=1}^{n}
\mathbf{X}_i
$$

satisfies

$$
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\xrightarrow{d}
\mathscr{N}
(\mathbf{0},\boldsymbol{\Sigma}).
$$

This result depends on the unknown population covariance matrix
\(\boldsymbol{\Sigma}\), which is generally not observable in practice.

---

# Slutsky's Theorem

Let

$$
X_n \xrightarrow{d} X
$$

and

$$
Y_n \xrightarrow{p} c,
$$

where \(c\) is a constant.

Then Slutsky's Theorem states that

$$
X_n + Y_n
\xrightarrow{d}
X + c,
$$

$$
X_nY_n
\xrightarrow{d}
cX,
$$

and, provided \(c\neq 0\),

$$
\frac{X_n}{Y_n}
\xrightarrow{d}
\frac{X}{c}.
$$

More generally, if a sequence of random matrices converges in probability to a fixed matrix, then that matrix may be replaced by its consistent estimator without affecting the asymptotic distribution.

---

# Consistency of the Sample Covariance Matrix

The sample covariance matrix is

$$
\mathbf{S}
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)
(\mathbf{X}_i-\widehat{\boldsymbol{\mu}}_n)^T.
$$

Under standard regularity conditions,

$$
\mathbf{S}
\xrightarrow{p}
\boldsymbol{\Sigma}.
$$

This means that the sample covariance matrix is a consistent estimator of the true population covariance matrix.

Therefore, as the sample size increases,

$$
\mathbf{S}
-
\boldsymbol{\Sigma}
\xrightarrow{p}
\mathbf{0}.
$$

---

# Applying Slutsky's Theorem

From the Multivariate Central Limit Theorem,

$$
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\xrightarrow{d}
\mathscr{N}
(\mathbf{0},\boldsymbol{\Sigma}).
$$

At the same time,

$$
\mathbf{S}
\xrightarrow{p}
\boldsymbol{\Sigma}.
$$

Since the covariance estimator converges in probability to the true covariance matrix, Slutsky's Theorem allows us to replace the unknown matrix \(\boldsymbol{\Sigma}\) by its consistent estimator \(\mathbf{S}\) without changing the asymptotic distribution.

Consequently,

$$
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\xrightarrow{d}
\mathscr{N}
(\mathbf{0},\mathbf{S}).
$$

Equivalently,

$$
\widehat{\boldsymbol{\mu}}_n
\xrightarrow{d}
\mathscr{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right).
$$

Thus, for sufficiently large sample sizes,

$$
\boxed{
\widehat{\boldsymbol{\mu}}_n
\sim
\mathscr{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right)
}
$$

which is the practical parametric model used in statistical inference.

---

# Interpretation

The Multivariate Central Limit Theorem provides the asymptotic Gaussian form of the sample mean but depends on the unknown covariance matrix \(\boldsymbol{\Sigma}\).

Because the sample covariance matrix \(\mathbf{S}\) converges in probability to \(\boldsymbol{\Sigma}\), Slutsky's Theorem guarantees that replacing the unknown population covariance by its empirical estimate does not alter the asymptotic distribution.

Therefore, in practical engineering and data-analysis applications, the sample mean can be modeled as

$$
\widehat{\boldsymbol{\mu}}_n
\sim
\mathscr{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right),
$$

allowing confidence regions, hypothesis tests, and uncertainty quantification to be constructed using only observable sample statistics.

# 1. Covariance Structure in Principal Component Space

Let the centered random vector be

$$
\widetilde{\mathbf{X}}_i
=
\mathbf{X}_i-\boldsymbol{\mu}
$$

with covariance matrix

$$
\boldsymbol{\Sigma}
=
\mathbb{E}
\left[
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
\right].
$$

Since \(\boldsymbol{\Sigma}\) is symmetric and positive semidefinite, it admits the spectral decomposition

$$
\boldsymbol{\Sigma}
=
\mathbf{P}
\mathbf{\Lambda}
\mathbf{P}^T,
$$

where

$$
\mathbf{P}
=
[\mathbf{p}_1,\mathbf{p}_2,\ldots,\mathbf{p}_m]
$$

is an orthogonal matrix satisfying

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I},
$$

and

$$
\mathbf{\Lambda}
=
\mathrm{diag}
(\lambda_1,\lambda_2,\ldots,\lambda_m).
$$

Define the principal component transformation

$$
\mathbf{Z}_i
=
\mathbf{P}^T
\widetilde{\mathbf{X}}_i.
$$

---

## Covariance of the Transformed Vector

The covariance matrix of \(\mathbf{Z}_i\) is

$$
\mathrm{Cov}(\mathbf{Z}_i)
=
\mathbb{E}
[\mathbf{Z}_i\mathbf{Z}_i^T].
$$

Substituting the transformation,

$$
=
\mathbb{E}
\left[
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)^T
\right].
$$

Using

$$
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)^T
=
\widetilde{\mathbf{X}}_i^T\mathbf{P},
$$

we obtain

$$
=
\mathbf{P}^T
\mathbb{E}
\left[
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
\right]
\mathbf{P}.
$$

Since

$$
\mathbb{E}
\left[
\widetilde{\mathbf{X}}_i
\widetilde{\mathbf{X}}_i^T
\right]
=
\boldsymbol{\Sigma},
$$

we have

$$
=
\mathbf{P}^T
\boldsymbol{\Sigma}
\mathbf{P}.
$$

Substituting the spectral decomposition,

$$
=
\mathbf{P}^T
(\mathbf{P}\mathbf{\Lambda}\mathbf{P}^T)
\mathbf{P}.
$$

Using orthogonality,

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I},
$$

gives

$$
=
\mathbf{\Lambda}.
$$

Hence,

$$
\boxed{
\mathbb{E}
[\mathbf{Z}_i\mathbf{Z}_i^T]
=
\mathbf{\Lambda}
}
$$

which is a diagonal matrix.

---

## Geometric and Statistical Interpretation

Since

$$
\mathbf{\Lambda}
=
\mathrm{diag}
(\lambda_1,\lambda_2,\ldots,\lambda_m),
$$

all off-diagonal entries are zero.

Therefore,

$$
\boxed{
\mathrm{Cov}(Z_{i,j},Z_{i,k})
=
0,
\qquad j\neq k
}
$$

meaning that distinct principal components are uncorrelated.

Geometrically:

- PCA rotates the coordinate system.
- The new axes coincide with the directions of maximum variance.
- The transformed coordinates are mutually orthogonal.

Statistically:

- Each principal component captures an independent source of variation.
- No linear redundancy remains between different coordinates.

\widehat{\mathbf{P}}_k
\perp
\widehat{\mathbf{P}}_{m-k},
$$

the retained and residual subspaces are orthogonal.

Hence,

$$
\left\|
\mathbf{x}_i-\widehat{\boldsymbol{\mu}}_n
\right\|^2
=
\left\|
\widehat{\mathbf{P}}_k
\mathbf{z}_{i,k}
+
\widehat{\mathbf{P}}_{m-k}
\mathbf{z}_{i,m-k}
\right\|^2.
$$

Expanding,

$$
=
\|\widehat{\mathbf{P}}_k\mathbf{z}_{i,k}\|^2
+
\|\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}\|^2
+
2
(\widehat{\mathbf{P}}_k\mathbf{z}_{i,k})^T
(\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}).
$$

The cross-term vanishes because of orthogonality:

$$
(\widehat{\mathbf{P}}_k)^T
\widehat{\mathbf{P}}_{m-k}
=
\mathbf{0}.
$$

Therefore,

$$
=
\|\mathbf{z}_{i,k}\|^2
+
\|\mathbf{z}_{i,m-k}\|^2.
$$

Hence,

$$
\boxed{
\|\mathbf{x}_i-\widehat{\boldsymbol{\mu}}_n\|^2
=
\|\mathbf{z}_{i,k}\|^2
+
\|\mathbf{z}_{i,m-k}\|^2
}
$$

which is the PCA Pythagorean identity.

# Hotelling's \(T^2\) Statistic vs Q Statistic (SPE)

## Hotelling's \(T^2\)

The \(T^2\) statistic measures variation within the retained PCA subspace:

$$
\boxed{
T^2
=
\mathbf{z}_{i,k}^T
\mathbf{\Lambda}_k^{-1}
\mathbf{z}_{i,k}
}
$$

It detects abnormal behavior that still follows the established correlation structure.

Examples:

- Extreme wind loads
- Heavy operational loading
- Temperature-driven operating changes

These events increase the magnitude of scores in the principal subspace.

Therefore,

$$
\boxed{
\text{Environmental / operational anomalies}
\rightarrow T^2
}
$$

---

## Q Statistic (Squared Prediction Error)

The Q statistic measures residual energy outside the retained PCA model:

$$
\boxed{
Q
=
\|\mathbf{e}_i\|^2
=
\|\mathbf{z}_{i,m-k}\|^2
}
$$

It detects new behavior not represented by the baseline covariance structure.

Examples:

- Fatigue cracks
- Structural fractures
- Sensor failures
- Changes in load transfer pathways

Such events create energy in directions excluded from the normal PCA model.

Therefore,

$$
\boxed{
\text{Structural damage or fatigue crack}
\rightarrow Q
}
$$

---

## Summary

$$
\boxed{
\text{Extreme operational load anomaly}
\rightarrow T^2
}
$$

because the event remains inside the learned PCA subspace.

$$
\boxed{
\text{Internal structural fracture}
\rightarrow Q
}
$$

because the damage introduces new variance directions outside the baseline model.

## **Part B: Numerical Verification**



***Scenario:** You are building a real-time diagnostics layer for a structural component monitored by $m = 4$ sensors. You have a baseline archive of $n = 3,000$ nominal snapshots. To establish operational boundaries, you must write an optimization pipeline that analyzes how Hotelling’s $T^2$ and the residual $Q$ statistic respond to different truncation boundaries $k \in \{1, 2, 3\}$.*

**Your Task:**
Complete the Python class method below inside Google Colab. The function must:

1. Standardize and center the incoming realizations.
2. Decompose the unbiased sample covariance matrix $\mathbf{S}$ using `np.linalg.eigh` and sort components in descending order ($\widehat{\lambda}_1 \ge \dots \ge \widehat{\lambda}_m$).
3. Loop through every possible subspace cutoff value $k \in \{1, 2, 3\}$, calculating the sample vector array for Hotelling’s $T^2$ ($T^2_i = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j}$) and the $Q$ statistic ($Q_i = \sum_{j=k+1}^m z_{i,j}^2$).
4. Compute the expected value (empirical mean) of $T^2$ and $Q$ across all samples for each $k$.
5. Return these metrics to verify your theoretical derivations in Part A.



````python
import numpy as np
import pandas as pd

class SHMPCAAnalyzer:

    def evaluate_subspace_statistics(self, X):
        """
        Evaluate Hotelling's T² and Q statistics for
        k = 1, 2, ..., m-1 principal components.

        Parameters
        ----------
        X : ndarray of shape (n_samples, n_features)

        Returns
        -------
        results : pd.DataFrame
        """

        # --------------------------------------------------
        # 1. Center the data
        # --------------------------------------------------
        X_centered = X - np.mean(X, axis=0)

        n, m = X_centered.shape

        # --------------------------------------------------
        # 2. Unbiased covariance matrix
        # --------------------------------------------------
        S = np.cov(X_centered, rowvar=False, bias=False)

        # --------------------------------------------------
        # 3. Eigen decomposition
        # --------------------------------------------------
        eigvals, eigvecs = np.linalg.eigh(S)

        # Sort in descending order
        idx = np.argsort(eigvals)[::-1]

        eigvals = eigvals[idx]
        eigvecs = eigvecs[:, idx]

        # --------------------------------------------------
        # PCA scores
        # --------------------------------------------------
        Z = X_centered @ eigvecs

        results = []

        # --------------------------------------------------
        # 4. Loop over k = 1,2,...,m-1
        # --------------------------------------------------
        for k in range(1, m):

            # -------- Hotelling T² --------
            T2 = np.sum(
                (Z[:, :k] ** 2) / eigvals[:k],
                axis=1
            )

            # -------- Q Statistic (SPE) --------
            Q = np.sum(
                Z[:, k:] ** 2,
                axis=1
            )

            results.append({
                "k": k,
                "Mean_T2": np.mean(T2),
                "Mean_Q": np.mean(Q)
            })

        return pd.DataFrame(results)
````

## **Part B: Parametric Factor Partition Engine & UI Architecture**

The objective is to implement an end-to-end data processing and diagnostic visualization pipeline that instantiates a parametric **Factor Analysis (FA)** engine with **Varimax rotation**. The engine must partition empirical observation metrics into a shared structural subspace and a distinct diagonal noise floor, mapping the results onto a highly structured, asymmetric grid interface.

### Develop a Mathematical Engine & Pre-processing Guardrails


Implement an isolated algorithm to structure and decompose empirical data structures according to the following guidelines:

1. **Z-Score Mapping Layer:** Accept an arbitrary data frame and isolate numeric observation columns. Standardize the observations into standard normal space $Z \in \mathbb{R}^{n \times m}$ applying an unbiased sample standard deviation ($ddof=1$). Build an explicit numerical guardrail that catches zero-variance channels ($\sigma_j = 0$) and caps them to a minimum boundary threshold of $10^{-15}$ to prevent divide-by-zero runtime exceptions.
2. **Dimensionality Constraint Layer:** Let $m$ represent the count of isolated sensor features, and $k$ represent the target count of latent physical factors. Program two validation rule exceptions:
* Reject processing if the operational snapshot count ($n$) is less than or equal to the channel count ($m$).
* Throw an descriptive error if the requested latent space dimension ($k$) is not strictly less than the available feature space dimension ($m$).


3. **Information Decomposition Layer:** Fit a Factor Analysis framework utilizing a Maximum Likelihood or Singular Value Decomposition (SVD) algorithm combined with an orthogonal Varimax rotation matrix ($T$). From the fitted parameters, compute the vector arrays for:
* **Communality Vector ($h^2 \in \mathbb{R}^m$):** The total shared variance captured across each channel.
* **Uniqueness Vector ($\varphi^2 \in \mathbb{R}^m$):** The remaining isolated sensor noise floor.

````python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis


def varimax(Phi, gamma=1.0, q=50, tol=1e-6):
    """
    Orthogonal Varimax rotation.
    """
    p, k = Phi.shape

    R = np.eye(k)
    d_old = 0

    for _ in range(q):

        Lambda = Phi @ R

        U, S, Vt = np.linalg.svd(
            Phi.T @ (
                Lambda**3
                - (gamma / p)
                * Lambda
                @ np.diag(np.sum(Lambda**2, axis=0))
            )
        )

        R = U @ Vt

        d = np.sum(S)

        if d_old != 0 and (d - d_old) < tol:
            break

        d_old = d

    return Phi @ R, R


def factor_analysis_pipeline(df, k):
    """
    Factor Analysis pipeline with:
        1. Z-score standardization
        2. Dimension validation
        3. Factor extraction
        4. Varimax rotation
        5. Communality & uniqueness computation
    """

    # --------------------------------------------------
    # 1. Isolate numeric columns
    # --------------------------------------------------

    X = df.select_dtypes(include=[np.number])

    if X.empty:
        raise ValueError(
            "No numeric observation columns found."
        )

    n, m = X.shape

    # --------------------------------------------------
    # 2. Dimension validation
    # --------------------------------------------------

    if n <= m:
        raise ValueError(
            f"Invalid dimensions: n={n}, m={m}. "
            "Requirement: n > m."
        )

    if k >= m:
        raise ValueError(
            f"Invalid latent dimension k={k}. "
            "Requirement: k < m."
        )

    # --------------------------------------------------
    # 3. Z-score mapping
    # --------------------------------------------------

    mu = X.mean(axis=0)

    sigma = X.std(axis=0, ddof=1)

    sigma = np.maximum(sigma, 1e-15)

    Z = (X - mu) / sigma

    Z = Z.to_numpy()

    # --------------------------------------------------
    # 4. Factor Analysis (Maximum Likelihood)
    # --------------------------------------------------

    fa = FactorAnalysis(
        n_components=k,
        random_state=42
    )

    fa.fit(Z)

    loadings = fa.components_.T

    # --------------------------------------------------
    # 5. Varimax rotation
    # --------------------------------------------------

    rotated_loadings, rotation_matrix = varimax(loadings)

    # --------------------------------------------------
    # 6. Communality vector
    # --------------------------------------------------

    communalities = np.sum(
        rotated_loadings**2,
        axis=1
    )

    # --------------------------------------------------
    # 7. Uniqueness vector
    # --------------------------------------------------

    uniqueness = 1.0 - communalities

    uniqueness = np.maximum(
        uniqueness,
        0.0
    )

    # --------------------------------------------------
    # Results
    # --------------------------------------------------

    results = {
        "Z": pd.DataFrame(
            Z,
            columns=X.columns
        ),

        "Rotated_Loadings": pd.DataFrame(
            rotated_loadings,
            index=X.columns,
            columns=[
                f"Factor_{i+1}"
                for i in range(k)
            ]
        ),

        "Communality_h2": pd.Series(
            communalities,
            index=X.columns
        ),

        "Uniqueness_phi2": pd.Series(
            uniqueness,
            index=X.columns
        ),

        "Rotation_Matrix": rotation_matrix
    }

    return results
````

````python
# Example dataset

np.random.seed(42)

df = pd.DataFrame(
    np.random.randn(3000, 5),
    columns=[
        "Sensor_1",
        "Sensor_2",
        "Sensor_3",
        "Sensor_4",
        "Sensor_5"
    ]
)

results = factor_analysis_pipeline(df, k=2)

print("Communality Vector (h²)")
print(results["Communality_h2"])

print("\nUniqueness Vector (φ²)")
print(results["Uniqueness_phi2"])

print("\nRotated Factor Loadings")
print(results["Rotated_Loadings"])
````

- \(Z \in \mathbb{R}^{n \times m}\) is the standardized observation matrix,
- \(R\) is the correlation matrix of the standardized observations,
- \(R^{-1}\) is the inverse (or pseudo-inverse) of the correlation matrix,
- \(\Lambda_{\text{rotated}}\) is the Varimax-rotated factor loading matrix,
- \(F \in \mathbb{R}^{n \times k}\) contains the estimated latent factor states.

This estimator provides the minimum mean squared error linear prediction of the hidden factors based on the observed measurements.

---

### Empirical Factor Variance

The empirical variance of each reconstructed factor is computed using the unbiased estimator:

$$
\widehat{\mathrm{Var}}(F_j)
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(F_{ij}-\bar{F}_j)^2
\qquad (j=1,\ldots,k)
$$

which corresponds to using `ddof=1` in NumPy.

---

### Average System Communality

The communality of sensor \(j\) is defined as

$$
h_j^2
=
\sum_{r=1}^{k}
\lambda_{jr}^2
$$

where \(\lambda_{jr}\) denotes the rotated loading of sensor \(j\) on factor \(r\).

The average system communality percentage is

$$
\boxed{
\text{Average System Communality (\%)}
=
\frac{1}{m}
\sum_{j=1}^{m}
h_j^2
\times 100
}
$$

This metric represents the average proportion of sensor variance explained by the common latent factors.

---

### Average System Uniqueness

The uniqueness of sensor \(j\) is defined as

$$
\varphi_j^2
=
1-h_j^2
$$

The average system uniqueness percentage is

$$
\boxed{
\text{Average System Uniqueness (\%)}
=
\frac{1}{m}
\sum_{j=1}^{m}
\varphi_j^2
\times 100
}
$$

This metric represents the average proportion of variance attributed to sensor-specific noise, local disturbances, and unexplained effects.

---

### Interpretation

A high average communality indicates that the sensor network is strongly governed by common latent physical mechanisms and that the factor model explains a large fraction of the observed variability.

A high average uniqueness indicates that a substantial portion of the observed variance remains sensor-specific and is not captured by the latent factor structure.

Since

$$
h_j^2+\varphi_j^2=1,
$$

the average communality percentage and average uniqueness percentage should approximately sum to \(100\%\), apart from minor numerical rounding effects.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def plot_pca_suite(loadings,
                   eigenvalues,
                   explained_variance_ratio,
                   mean_t2,
                   mean_q):

    cumulative_variance = np.cumsum(explained_variance_ratio) * 100
    explained_pct = explained_variance_ratio * 100
    residual_pct = 100 - cumulative_variance

    fig, axes = plt.subplots(
        2, 3,
        figsize=(18, 10)
    )

    # --------------------------------------------------
    # Panel 1,1 : Loading Heatmap
    # --------------------------------------------------
    ax = axes[0,0]

    im = ax.imshow(
        np.abs(loadings),
        aspect='auto'
    )

    ax.set_title("Feature Loadings |P(j,r)|")
    ax.set_xlabel("Principal Components")
    ax.set_ylabel("Sensors")

    plt.colorbar(im, ax=ax)

    # --------------------------------------------------
    # Panel 1,2 : Eigenvalues
    # --------------------------------------------------
    ax = axes[0,1]

    ax.bar(
        np.arange(1, len(eigenvalues)+1),
        eigenvalues
    )

    ax.set_title("Eigenvalue Magnitudes")
    ax.set_xlabel("Component")
    ax.set_ylabel("Eigenvalue")

    # --------------------------------------------------
    # Panel 1,3 : Explained Variance
    # --------------------------------------------------
    ax = axes[0,2]

    x = np.arange(1, len(eigenvalues)+1)

    ax.bar(
        x,
        explained_pct,
        label="Explained Variance (%)"
    )

    ax.plot(
        x,
        cumulative_variance,
        '--o',
        linewidth=2,
        label="Cumulative Variance (%)"
    )

    ax.set_title("Variance Capture")
    ax.legend()

    # --------------------------------------------------
    # Panel 2,1 : Residual Variance
    # --------------------------------------------------
    ax = axes[1,0]

    ax.bar(
        x,
        residual_pct
    )

    ax.set_title("Residual Variance (%)")
    ax.set_xlabel("k")
    ax.set_ylabel("%")

    # --------------------------------------------------
    # Panel 2,2 : Mean T²
    # --------------------------------------------------
    ax = axes[1,1]

    ax.plot(
        np.arange(1, len(mean_t2)+1),
        mean_t2,
        'o-'
    )

    ax.set_title("Mean Hotelling T²")
    ax.set_xlabel("Subspace Size k")

    # --------------------------------------------------
    # Panel 2,3 : Mean Q
    # --------------------------------------------------
    ax = axes[1,2]

    ax.plot(
        np.arange(1, len(mean_q)+1),
        mean_q,
        'o-'
    )

    ax.set_title("Mean Q Statistic (SPE)")
    ax.set_xlabel("Subspace Size k")

    plt.tight_layout()
    plt.show()
````

## Interpretation of the Rotated Factor Analysis Loadings Matrix

### 1. How Varimax Rotation Creates a Simple Structure

In Principal Component Analysis (PCA), the principal components are obtained by solving the eigenvalue problem

$$
\Sigma = P\Lambda P^T,
$$

where

$$
\lambda_1 \geq \lambda_2 \geq \cdots \geq \lambda_m.
$$

The first principal component is forced to capture the maximum possible variance, the second captures the largest remaining variance, and so forth.

As a consequence, PCA components often contain contributions from many sensors simultaneously. A typical PCA loading vector may appear as

$$
PC_1 =
0.48S_1 +
0.52S_2 +
0.44S_3 +
0.55S_4.
$$

Although mathematically optimal for variance capture, such a component mixes several physical sources together.

---

Factor Analysis introduces a loading matrix

$$
\Lambda
=
[\lambda_{jr}]
$$

and then applies an orthogonal rotation matrix

$$
T
$$

to obtain

$$
\Lambda_{\text{rotated}}
=
\Lambda T.
$$

Varimax rotation chooses \(T\) so that each factor contains:

- A few very large loadings,
- Many near-zero loadings.

This process is called **simple structure**.

Instead of every factor sharing variance across all sensors, the rotation encourages clean separation of sensor groups.

For example, after Varimax rotation the loading matrix may resemble

$$
\begin{array}{c|cc}
 & F_1 & F_2\\
\hline
S_1 & 0.91 & 0.05\\
S_2 & 0.88 & 0.02\\
S_3 & 0.03 & 0.93\\
S_4 & 0.07 & 0.10
\end{array}
$$

In this structure:

- `Sensor_1` and `Sensor_2` belong primarily to Factor 1.
- `Sensor_3` belongs primarily to Factor 2.
- `Sensor_4` contributes weakly to both factors.

The factors become physically interpretable rather than purely mathematical variance directions.

---

### 2. Why the Rotated FA Heatmap Is Easier to Troubleshoot

From a plant operator's perspective, the objective is not merely to maximize variance explained. The objective is to identify which physical subsystem is responsible for abnormal behavior.

A raw PCA eigenvector matrix often contains mixed loadings across many sensors:

$$
PC_1
=
0.48S_1
+
0.52S_2
+
0.44S_3
+
0.55S_4.
$$

If an anomaly appears in \(PC_1\), it is difficult to determine which sensor group or physical mechanism caused the event because all sensors contribute simultaneously.

The operator must interpret a mathematical combination of multiple signals rather than a distinct physical subsystem.

---

In contrast, the rotated FA heatmap produces localized factor groupings.

For example:

- Factor 1 may represent a structural load pathway monitored by `Sensor_1` and `Sensor_2`.
- Factor 2 may represent a different subsystem monitored primarily by `Sensor_3`.

If an abnormal factor score is observed:

- A rise in Factor 1 immediately directs attention toward the subsystem associated with `Sensor_1` and `Sensor_2`.
- A rise in Factor 2 immediately points toward the subsystem associated with `Sensor_3`.

This greatly simplifies fault isolation.

---

### Engineering Significance

The PCA loading matrix answers:

> "Which mathematical directions explain the most variance?"

The rotated Factor Analysis loading matrix answers:

> "Which groups of sensors are responding to the same underlying physical mechanism?"

Therefore, during structural failures, equipment degradation, or process disturbances, the Varimax-rotated Factor Analysis heatmap is substantially easier to interpret because it produces clean, physically meaningful sensor clusters rather than mixed variance directions. This makes troubleshooting, root-cause analysis, and maintenance decisions much faster and more reliable.

## Interpretation of the Mean Q Statistic (SPE) Curve

### 1. Behavior of the Mean Q Statistic as \(k\) Increases

The Q Statistic (Squared Prediction Error) measures the residual energy remaining outside the retained PCA subspace.

Mathematically,

$$
Q_i
=
\sum_{j=k+1}^{m}
z_{i,j}^{2},
$$

where the retained components are excluded and only the discarded residual coordinates contribute.

---

### Transition from \(k=1\) to \(k=2\)

When only one principal component is retained (\(k=1\)), a substantial portion of the system variance remains in the residual space.

The Mean Q Statistic is therefore relatively large.

After increasing the cutoff to

$$
k=2,
$$

the second major principal component is incorporated into the retained subspace.

Since the second component contains a large amount of structured system variance, a significant amount of residual energy is removed.

Consequently,

$$
\boxed{
\text{Mean }Q
\text{ decreases sharply from }
k=1
\text{ to }
k=2
}
$$

This sharp drop indicates that the second component captures important physical information rather than noise.

---

### Transition from \(k=2\) to \(k=3\)

When the cutoff is increased further to

$$
k=3,
$$

the Mean Q Statistic decreases only slightly.

The residual curve begins to flatten because most of the meaningful system variance has already been captured by the first two principal components.

Therefore,

$$
\boxed{
\text{Mean }Q
\text{ shows only a small decrease from }
k=2
\text{ to }
k=3
}
$$

This behavior produces a characteristic elbow in the curve.

---

### 2. Why the Elbow Identifies \(k=2\) as the True Physical Dimensionality

A PCA model seeks the smallest subspace that captures the dominant physical structure of the system.

If the first two principal components explain nearly all meaningful covariance relationships, then

$$
\mathbf{x}
\approx
\mathbf{P}_2\mathbf{z}_2.
$$

Adding additional components after \(k=2\) contributes little new information.

The Q curve therefore exhibits:

- A large reduction when moving from \(k=1\) to \(k=2\),
- A much smaller reduction when moving from \(k=2\) to \(k=3\).

This creates an elbow shape.

Graphically:

```text
Q
│
│\
│ \
│  \
│   \____
│        \__
└────────────── k
    1    2   3
```

The elbow occurs at the point where the major structured variance has been captured and only weak residual variation remains.

Therefore,

$$
\boxed{
k=2
}
$$

is interpreted as the intrinsic dimensionality of the underlying physical system.

In other words, the asset behaves as if it is governed primarily by two independent physical mechanisms.

---

### 3. Consequences of Choosing \(k=3\)

When the cutoff is increased beyond the true system dimensionality,

$$
k > 2,
$$

the PCA model begins absorbing residual variation that should remain outside the principal subspace.

This additional variation often consists of:

- Measurement noise,
- Sensor-specific disturbances,
- Environmental fluctuations,
- Numerical uncertainty,
- Instrument drift.

Consequently, selecting

$$
k=3
$$

forces part of the residual noise space into the retained PCA model.

Mathematically,

$$
\mathbf{x}
=
\mathbf{P}_2\mathbf{z}_2
+
\mathbf{P}_{\text{noise}}
\mathbf{z}_{\text{noise}}
$$

and the third component begins representing portions of

$$
\mathbf{P}_{\text{noise}}
\mathbf{z}_{\text{noise}}.
$$

rather than meaningful structural behavior.

---

### Engineering Interpretation

Choosing \(k=3\) causes the monitoring system to treat some noise as if it were legitimate system dynamics.

This can produce several undesirable effects:

1. Reduced sensitivity of the Q Statistic to structural faults.
2. Increased model complexity without meaningful information gain.
3. Greater risk of masking emerging damage signatures.
4. Increased likelihood of overfitting the baseline data.

Therefore, the elbow in the Mean Q Statistic curve provides strong evidence that

$$
\boxed{
k=2
}
$$

is the most appropriate truncation boundary because it captures the true physical structure while leaving sensor noise and unexplained disturbances in the residual monitoring space.

## Comparison of PCA-Based and Factor Analysis-Based Predictive Maintenance Strategies

### 1. PCA Monitoring Strategy

The PCA approach decomposes the observation space into:

- A retained principal subspace monitored using Hotelling's \(T^2\),
- A residual subspace monitored using the \(Q\) statistic (SPE).

The monitoring statistics are

$$
T^2
=
\mathbf{z}_k^T
\mathbf{\Lambda}_k^{-1}
\mathbf{z}_k
$$

and

$$
Q
=
\|\mathbf{e}\|^2.
$$

The \(T^2\) statistic measures how far an observation moves within the learned principal operating space, while the \(Q\) statistic measures energy that falls outside the PCA model.

This strategy is highly effective for detecting:

- Changes in operating conditions,
- Global process shifts,
- Environmental loading changes,
- Multivariate deviations from the baseline operating region.

Because PCA maximizes total variance, it captures dominant system behavior using a compact set of principal components.

---

### 2. Factor Analysis Monitoring Strategy

Factor Analysis models the system using latent physical factors:

$$
\mathbf{x}
=
\boldsymbol{\Lambda}\mathbf{f}
+
\boldsymbol{\epsilon}.
$$

The monitoring variables become the rotated factor scores

$$
\mathbf{f}
=
\boldsymbol{\Lambda}^T
\mathbf{R}^{-1}
\mathbf{z}.
$$

Each factor score represents a physically interpretable latent mechanism extracted from the sensor network.

Instead of monitoring variance directions, the FA framework monitors:

- Structural load pathways,
- Common physical behaviors,
- Latent system states.

The rotation process produces a simple structure where groups of sensors are associated with specific factors, making fault isolation significantly easier.

---

## Robustness to Sensor Calibration Loss or Electrical Shorts

Suppose a single sensor experiences:

- Calibration drift,
- Electrical interference,
- Wiring faults,
- Amplifier saturation,
- Complete sensor degradation.

Such failures generate variance that is localized to one channel rather than shared across the entire sensor network.

Factor Analysis explicitly separates total variance into

$$
1
=
h_j^2
+
\varphi_j^2,
$$

where:

- \(h_j^2\) = communality (shared structural variance),
- \(\varphi_j^2\) = uniqueness (sensor-specific variance).

A malfunctioning sensor typically exhibits

$$
\varphi_j^2 \rightarrow 1.
$$

This means that most of its variance is isolated from the common latent factors.

Consequently, Factor Analysis naturally identifies problematic sensors through elevated uniqueness values.

For example, if

$$
\varphi_4^2 = 0.98,
$$

then approximately \(98\%\) of the variance observed by Sensor 4 is sensor-specific and contributes very little information about the actual structural state.

The uniqueness profile therefore acts as a built-in sensor health indicator.

---

## Why PCA Is More Vulnerable

PCA does not distinguish between:

- Shared structural variance,
- Sensor-specific noise.

Instead, it maximizes total variance:

$$
\Sigma
=
P\Lambda P^T.
$$

A faulty sensor with large variance can therefore increase the covariance matrix and artificially inflate one or more eigenvalues.

As a result:

- Noise may be absorbed into the principal subspace.
- Hotelling's \(T^2\) thresholds may become distorted.
- Principal components may partially represent sensor faults rather than true system dynamics.

The PCA model may incorrectly interpret sensor degradation as meaningful process variation.

---

## Engineering Conclusion

The PCA strategy provides an excellent framework for detecting global operational changes through the combined use of:

- Hotelling's \(T^2\),
- Residual \(Q\) statistics.

However, PCA treats all variance equally and cannot inherently distinguish physical behavior from localized sensor faults.

Factor Analysis is generally more robust against calibration loss or electrical shorts because the uniqueness metric

$$
\varphi_j^2
$$

explicitly quantifies sensor-specific variance.

A faulty sensor naturally produces an elevated uniqueness value while contributing little to the common latent factors.

Therefore,

$$
\boxed{
\text{Factor Analysis is more robust to single-sensor failures}
}
$$

because the Sensor Uniqueness Noise Floor directly identifies localized measurement problems before they contaminate the latent structural state estimates.

In practical predictive maintenance systems, PCA is highly effective for detecting overall process anomalies, while Factor Analysis provides superior robustness and fault isolation when individual sensors degrade or fail. The most reliable industrial monitoring architectures often combine both approaches: PCA for global anomaly detection and Factor Analysis for sensor-level diagnostic interpretation.
