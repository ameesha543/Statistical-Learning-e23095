### 1. Maximum Likelihood Estimator for the Covariance Matrix

#### Derivation of Bias in $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$

Let's formally derive why the raw Maximum Likelihood Estimator for the covariance matrix, $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$, is inherently **biased** in finite samples.

First, recall the definitions:
*   $\mathbf{X}_1,\mathbf{X}_2,\cdots, \mathbf{X}_n\in \mathbb{R}^m$ are i.i.d. random variables.
*   $\boldsymbol{\mu}=\mathbb{E}(\mathbf{X}_i)$ and $\boldsymbol{\Sigma}=\mathrm{Var}(\mathbf{X}_i)$.
*   The sample mean is $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_{i}$.

The Maximum Likelihood Estimator for the covariance matrix is given by:
$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

To show the bias, we need to compute $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]$.

Let's start by expanding the term $(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$:

$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = \mathbf{X}_i - \boldsymbol{\mu} - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$

Let $\mathbf{Z}_i = \mathbf{X}_i - \boldsymbol{\mu}$ and $\overline{\mathbf{Z}} = \widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu} = \frac{1}{n} \sum_{i=1}^n \mathbf{Z}_i$.

Then, $\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = \mathbf{Z}_i - \overline{\mathbf{Z}}$.

So, $(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = (\mathbf{Z}_i - \overline{\mathbf{Z}})(\mathbf{Z}_i - \overline{\mathbf{Z}})^T$
$= \mathbf{Z}_i \mathbf{Z}_i^T - \mathbf{Z}_i \overline{\mathbf{Z}}^T - \overline{\mathbf{Z}} \mathbf{Z}_i^T + \overline{\mathbf{Z}} \overline{\mathbf{Z}}^T$

Now, sum over $i$ from $1$ to $n$:

$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{Z}_i - \overline{\mathbf{Z}})(\mathbf{Z}_i - \overline{\mathbf{Z}})^T$
$= \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - \sum_{i=1}^n \mathbf{Z}_i \overline{\mathbf{Z}}^T - \sum_{i=1}^n \overline{\mathbf{Z}} \mathbf{Z}_i^T + \sum_{i=1}^n \overline{\mathbf{Z}} \overline{\mathbf{Z}}^T$

Notice that $\sum_{i=1}^n \mathbf{Z}_i = n\overline{\mathbf{Z}}$. Substituting this:

$= \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - (n\overline{\mathbf{Z}}) \overline{\mathbf{Z}}^T - \overline{\mathbf{Z}} (n\overline{\mathbf{Z}})^T + n \overline{\mathbf{Z}} \overline{\mathbf{Z}}^T$
$= \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - n\overline{\mathbf{Z}} \overline{\mathbf{Z}}^T - n\overline{\mathbf{Z}} \overline{\mathbf{Z}}^T + n \overline{\mathbf{Z}} \overline{\mathbf{Z}}^T$
$= \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - n\overline{\mathbf{Z}} \overline{\mathbf{Z}}^T$

So, $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \left( \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - n\overline{\mathbf{Z}} \overline{\mathbf{Z}}^T \right) = \frac{1}{n} \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - \overline{\mathbf{Z}} \overline{\mathbf{Z}}^T$

Now, let's take the expectation:

$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \mathbb{E}\left[ \frac{1}{n} \sum_{i=1}^n \mathbf{Z}_i \mathbf{Z}_i^T - \overline{\mathbf{Z}} \overline{\mathbf{Z}}^T \right]$
$= \frac{1}{n} \sum_{i=1}^n \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] - \mathbb{E}[\overline{\mathbf{Z}} \overline{\mathbf{Z}}^T]$

We know that $\mathbb{E}[\mathbf{Z}_i] = \mathbb{E}[\mathbf{X}_i - \boldsymbol{\mu}] = \mathbf{0}$.
And $\mathrm{Var}(\mathbf{X}_i) = \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \boldsymbol{\Sigma}$.

For the second term, $\mathbb{E}[\overline{\mathbf{Z}} \overline{\mathbf{Z}}^T] = \mathrm{Var}(\overline{\mathbf{Z}})$.
Since $\overline{\mathbf{Z}} = \frac{1}{n} \sum_{i=1}^n \mathbf{Z}_i$ and the $\mathbf{Z}_i$ are i.i.d.,
$\mathrm{Var}(\overline{\mathbf{Z}}) = \mathrm{Var}\left(\frac{1}{n} \sum_{i=1}^n \mathbf{Z}_i\right) = \frac{1}{n^2} \sum_{i=1}^n \mathrm{Var}(\mathbf{Z}_i) = \frac{1}{n^2} \sum_{i=1}^n \boldsymbol{\Sigma} = \frac{1}{n^2} (n\boldsymbol{\Sigma}) = \frac{1}{n} \boldsymbol{\Sigma}$.

Substituting these back into the expectation:

$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \boldsymbol{\Sigma} - \frac{1}{n} \boldsymbol{\Sigma}$
$= \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n} \boldsymbol{\Sigma}$
$= \boldsymbol{\Sigma} - \frac{1}{n} \boldsymbol{\Sigma}$
$= \left(1 - \frac{1}{n}\right) \boldsymbol{\Sigma}$
$= \frac{n-1}{n} \boldsymbol{\Sigma}$

Since $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n-1}{n}\boldsymbol{\Sigma} \neq \boldsymbol{\Sigma}$ for finite $n$, the Maximum Likelihood Estimator for the covariance matrix is **biased**.

#### Bessel's Correction for Unbiased Estimator

To obtain an unbiased estimator, we need to adjust $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ such that its expectation is $\boldsymbol{\Sigma}$.

From the previous derivation, we have $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n-1}{n}\boldsymbol{\Sigma}$.

We want an estimator $\mathbf{S}$ such that $\mathbb{E}[\mathbf{S}] = \boldsymbol{\Sigma}$.

Let $\mathbf{S} = c \cdot \widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ for some constant $c$.
Then, $\mathbb{E}[\mathbf{S}] = \mathbb{E}\left[ c \cdot \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} \right] = c \cdot \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = c \cdot \frac{n-1}{n} \boldsymbol{\Sigma}$.

For $\mathbf{S}$ to be unbiased, we must have $c \cdot \frac{n-1}{n} \boldsymbol{\Sigma} = \boldsymbol{\Sigma}$.
This implies $c \cdot \frac{n-1}{n} = 1$, so $c = \frac{n}{n-1}$.

Therefore, the unbiased estimator $\mathbf{S}$ is:
$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right)$$
$$\mathbf{S} = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

This is precisely Bessel's correction, which yields the unique, unbiased estimator $\mathbf{S}$ for the population covariance matrix.

### 2. Errors in Continuous Structural Health Monitoring

Under the null hypothesis $H_0$ that a structural asset remains completely healthy, a chosen test statistic tracks a deterministic parametric distribution.

#### Definition of Type I Error ($\alpha$) and Type II Error ($\beta$)

In the context of continuous structural health monitoring:

*   **Type I Error ($\alpha$) - False Positive (False Alarm):** This occurs when the monitoring system *incorrectly rejects the null hypothesis* ($H_0$), meaning it signals a structural anomaly or degradation (e.g., crack formation, excessive vibration) when, in reality, the asset is still healthy and operating within its normal parameters. In practical terms, it's a false alarm that suggests a problem that doesn't exist, potentially leading to unnecessary inspections, maintenance, or even costly shutdowns.

*   **Type II Error ($\beta$) - False Negative (Missed Detection):** This occurs when the monitoring system *incorrectly fails to reject the null hypothesis* ($H_0$), meaning it concludes that the asset is healthy when, in reality, a structural anomaly or degradation *has* occurred. In practical terms, it's a missed detection where a real problem goes unnoticed, which could have catastrophic consequences, such as structural failure, safety hazards, and significant financial losses if the issue escalates.

#### Consequence of a Conservative Significance Threshold ($\alpha = 0.0001$)

If an engineer sets an ultra-conservative significance threshold (e.g., $\alpha = 0.0001$) to prevent false alarms from shutting down an industrial plant, this decision has a direct and significant mathematical consequence on the statistical power of the diagnostic layer:

*   **Impact on Statistical Power:** Statistical power (often denoted as $1 - \beta$) is the probability of correctly rejecting the null hypothesis when it is false (i.e., detecting a real structural anomaly when one exists). There is an inverse relationship between Type I error ($\alpha$) and Type II error ($\beta$). When $\alpha$ is decreased (made more conservative), $\beta$ typically increases. Consequently, the statistical power ($1 - \beta$) **decreases**. This means the diagnostic system becomes less likely to detect actual structural problems, even when they are present. The engineer prioritizes avoiding false alarms at the cost of increasing the risk of missed detections.

*   **Modification of the "Healthy Operation" Confidence Ellipsoid:** The confidence ellipsoid represents the region in the multidimensional space of the test statistics where the structural asset is considered to be operating healthily under $H_0$. The significance level $\alpha$ defines the boundary of this ellipsoid. When $\alpha$ is set to an ultra-conservative value like $0.0001$, the system requires much stronger evidence to conclude that an observation falls *outside* the healthy region. Mathematically, this means the **geometric volume of your "healthy operation" confidence ellipsoid will significantly increase (expand)**. The critical region (the area outside the ellipsoid where you would reject $H_0$) shrinks dramatically. This expanded ellipsoid implies that a wider range of observed deviations in the test statistics will still be classified as "healthy operation," making the detection of true anomalies much harder and demanding very extreme, unambiguous signals of damage before an alarm is triggered.
