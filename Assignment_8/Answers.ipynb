# Statistical Mechanics of the Quantum Harmonic Oscillator

Consider a quantum harmonic oscillator with energy eigenvalues

$$E_j = \hbar\omega\left(j + \frac{1}{2}\right), \qquad j = 0, 1, 2, \ldots$$

Define

$$\beta = \frac{1}{k_B T}$$

and introduce the dimensionless quantity

$$x = \beta\hbar\omega = \frac{\hbar\omega}{k_B T}.$$

---

## 1. Partition Function

The canonical partition function is

$$Z(T) = \sum_{j=0}^{\infty} e^{-\beta E_j}.$$

Substituting the energy eigenvalues,

$$Z = \sum_{j=0}^{\infty} e^{-\beta\hbar\omega\left(j + \frac{1}{2}\right)}.$$

Separate the zero-point energy term:

$$Z = e^{-\frac{1}{2}\beta\hbar\omega} \sum_{j=0}^{\infty} e^{-j\beta\hbar\omega}.$$

The remaining summation is a geometric series:

$$1 + e^{-\beta\hbar\omega} + e^{-2\beta\hbar\omega} + \cdots$$

with common ratio

$$r = e^{-\beta\hbar\omega}.$$

Since $T > 0$, $\beta > 0$, and therefore

$$0 < e^{-\beta\hbar\omega} < 1.$$

Thus, the geometric series converges. Using

$$\sum_{j=0}^{\infty} r^j = \frac{1}{1 - r}, \qquad |r| < 1,$$

we obtain

$$\boxed{Z(T) = \frac{e^{-\frac{\beta\hbar\omega}{2}}}{1 - e^{-\beta\hbar\omega}}}$$

or, in terms of temperature,

$$\boxed{Z(T) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}}.$$

The partition function can also be written in hyperbolic form. Since

$$2\sinh\left(\frac{x}{2}\right) = e^{x/2} - e^{-x/2},$$

we have

$$Z = \frac{1}{2\sinh(x/2)}.$$

Therefore,

$$\boxed{Z = \frac{1}{2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)}}.$$

---

## 2. Thermodynamic Potentials

### 2.1 Helmholtz Free Energy

The Helmholtz free energy is

$$F = -k_B T \ln Z.$$

From the partition function,

$$\ln Z = -\frac{\beta\hbar\omega}{2} - \ln\left(1 - e^{-\beta\hbar\omega}\right).$$

Therefore,

$$F = -k_B T \left[ -\frac{\beta\hbar\omega}{2} - \ln\left(1 - e^{-\beta\hbar\omega}\right) \right].$$

Since $k_B T \beta = 1$,

$$\boxed{F(T) = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)}.$$

Using the hyperbolic form of the partition function,

$$\ln Z = -\ln\left[2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)\right],$$

so the free energy may also be written as

$$\boxed{F(T) = k_B T \ln\left[2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)\right]}.$$

---

### 2.2 Internal Energy

The internal energy is

$$U = -\frac{\partial \ln Z}{\partial \beta}.$$

Using

$$\ln Z = -\frac{\beta\hbar\omega}{2} - \ln\left(1 - e^{-\beta\hbar\omega}\right),$$

differentiate the first term:

$$\frac{\partial}{\partial \beta}\left(-\frac{\beta\hbar\omega}{2}\right) = -\frac{\hbar\omega}{2}.$$

For the second term,

$$\frac{\partial}{\partial \beta}\left[-\ln\left(1 - e^{-\beta\hbar\omega}\right)\right] = -\frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}.$$

Hence,

$$\frac{\partial \ln Z}{\partial \beta} = -\frac{\hbar\omega}{2} - \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}.$$

Therefore,

$$U = \frac{\hbar\omega}{2} + \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}.$$

Multiplying the second fraction by $e^{\beta\hbar\omega}/e^{\beta\hbar\omega}$,

$$\boxed{U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}}.$$

The first term,

$$\frac{\hbar\omega}{2},$$

is the zero-point energy. The second term represents the thermal excitation energy.

Using the identity

$$\coth\left(\frac{x}{2}\right) = \frac{e^x + 1}{e^x - 1},$$

the internal energy may also be written as

$$\boxed{U(T) = \frac{\hbar\omega}{2} \coth\left(\frac{\hbar\omega}{2k_B T}\right)}.$$

---

## 3. Entropy

The entropy is

$$S = -\left(\frac{\partial F}{\partial T}\right)_V.$$

Let

$$x = \frac{\hbar\omega}{k_B T}.$$

The free energy is

$$F = \frac{\hbar\omega}{2} + k_B T \ln(1 - e^{-x}).$$

The zero-point energy is independent of temperature, so its derivative is zero. Therefore,

$$\frac{\partial F}{\partial T} = k_B \ln(1 - e^{-x}) + k_B T \frac{\partial}{\partial T} \ln(1 - e^{-x}).$$

Since

$$x = \frac{\hbar\omega}{k_B T},$$

we have

$$\frac{dx}{dT} = -\frac{x}{T}.$$

Also,

$$\frac{d}{dx}\ln(1 - e^{-x}) = \frac{e^{-x}}{1 - e^{-x}} = \frac{1}{e^x - 1}.$$

Thus,

$$\frac{\partial}{\partial T} \ln(1 - e^{-x}) = -\frac{x}{T(e^x - 1)}.$$

Substituting,

$$\frac{\partial F}{\partial T} = k_B \ln(1 - e^{-x}) - \frac{k_B x}{e^x - 1}.$$

Therefore,

$$\boxed{S(T) = k_B \left[ \frac{x}{e^x - 1} - \ln(1 - e^{-x}) \right]}$$

where

$$x = \frac{\hbar\omega}{k_B T}.$$

Hence,

$$\boxed{S(T) = k_B \left[ \frac{\frac{\hbar\omega}{k_B T}}{e^{\frac{\hbar\omega}{k_B T}} - 1} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]}.$$

An equivalent hyperbolic form is

$$\boxed{S = k_B \left[ \frac{x}{2}\coth\left(\frac{x}{2}\right) - \ln\left(2\sinh\frac{x}{2}\right) \right]}.$$

---

### Verification of $U = F + TS$

Using

$$F = \frac{\hbar\omega}{2} + k_B T \ln(1 - e^{-x})$$

and

$$TS = k_B T \left[ \frac{x}{e^x - 1} - \ln(1 - e^{-x}) \right],$$

we get

$$F + TS = \frac{\hbar\omega}{2} + k_B T \ln(1 - e^{-x}) + \frac{k_B T x}{e^x - 1} - k_B T \ln(1 - e^{-x}).$$

The logarithmic terms cancel:

$$F + TS = \frac{\hbar\omega}{2} + \frac{k_B T x}{e^x - 1}.$$

Since

$$k_B T x = \hbar\omega,$$

we obtain

$$F + TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^x - 1}.$$

This is exactly the internal energy:

$$\boxed{U = F + TS}.$$

---

## 4. Heat Capacity and Hyperbolic Form

The constant-volume heat capacity is

$$C_V = \left(\frac{\partial U}{\partial T}\right)_V.$$

The internal energy is

$$U = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^x - 1},$$

where

$$x = \frac{\hbar\omega}{k_B T}.$$

The zero-point energy does not depend on temperature, so it does not contribute to the heat capacity.

Therefore,

$$C_V = \hbar\omega \frac{d}{dT}\left(\frac{1}{e^x - 1}\right).$$

First,

$$\frac{d}{dx}\left(\frac{1}{e^x - 1}\right) = -\frac{e^x}{(e^x - 1)^2}.$$

Also,

$$\frac{dx}{dT} = -\frac{x}{T}.$$

Thus,

$$C_V = \hbar\omega \left[-\frac{e^x}{(e^x - 1)^2}\right]\left(-\frac{x}{T}\right).$$

Hence,

$$C_V = \frac{\hbar\omega x}{T} \frac{e^x}{(e^x - 1)^2}.$$

Since

$$\hbar\omega = x k_B T,$$

we obtain

$$\boxed{C_V = k_B x^2 \frac{e^x}{(e^x - 1)^2}}.$$

Therefore,

$$\boxed{C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2}}.$$

---

### Conversion to Hyperbolic Form

Using

$$\sinh\left(\frac{x}{2}\right) = \frac{e^{x/2} - e^{-x/2}}{2},$$

squaring gives

$$\sinh^2\left(\frac{x}{2}\right) = \frac{e^x - 2 + e^{-x}}{4}.$$

Factor out $e^{-x}$:

$$e^x - 2 + e^{-x} = e^{-x}(e^x - 1)^2.$$

Therefore,

$$\sinh^2\left(\frac{x}{2}\right) = \frac{(e^x - 1)^2}{4e^x}.$$

Taking the reciprocal,

$$\frac{1}{\sinh^2(x/2)} = \frac{4e^x}{(e^x - 1)^2}.$$

Thus,

$$\frac{e^x}{(e^x - 1)^2} = \frac{1}{4\sinh^2(x/2)}.$$

Substituting this into the heat-capacity expression,

$$C_V = k_B x^2 \frac{1}{4\sinh^2(x/2)}.$$

Therefore,

$$C_V = k_B \left(\frac{x}{2}\right)^2 \frac{1}{\sinh^2(x/2)}.$$

Since

$$x = \frac{\hbar\omega}{k_B T},$$

the final result is

$$\boxed{C_V = k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}}.$$

---

## 5. High-Temperature Asymptotic Behaviour

In the high-temperature limit,

$$k_B T \gg \hbar\omega.$$

Therefore,

$$x = \frac{\hbar\omega}{k_B T} \ll 1.$$

For small $y$,

$$\sinh y \approx y.$$

Taking

$$y = \frac{\hbar\omega}{2k_B T},$$

the heat capacity becomes

$$C_V \approx k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\left(\frac{\hbar\omega}{2k_B T}\right)^2}.$$

Therefore,

$$\boxed{C_V \rightarrow k_B} \qquad \text{as } T \rightarrow \infty.$$

More precisely,

$$C_V \approx k_B \left[ 1 - \frac{1}{12}\left(\frac{\hbar\omega}{k_B T}\right)^2 + \cdots \right].$$

The internal energy at high temperature approaches

$$U \approx k_B T.$$

This agrees with the classical equipartition theorem. A classical harmonic oscillator has two quadratic energy terms:

$$E = \frac{p^2}{2m} + \frac{1}{2}m\omega^2 q^2.$$

Each quadratic term contributes

$$\frac{1}{2}k_B T$$

to the average energy. Therefore,

$$U_{\text{classical}} = \frac{1}{2}k_B T + \frac{1}{2}k_B T = k_B T,$$

and consequently,

$$\boxed{C_{V,\text{classical}} = k_B}.$$

Thus, the quantum harmonic oscillator approaches the classical equipartition result in the high-temperature limit.

---

## Final Results

$$\boxed{Z = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}} = \frac{1}{2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)}}$$

$$\boxed{F = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)}$$

$$\boxed{U = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}}$$

$$\boxed{S = k_B \left[ \frac{\frac{\hbar\omega}{k_B T}}{e^{\frac{\hbar\omega}{k_B T}} - 1} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]}$$

$$\boxed{C_V = k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}}$$

and in the high-temperature limit,

$$\boxed{C_V \rightarrow k_B}.$$


# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# Maximum Entropy Distribution with Fixed Mean and Variance

Let $X$ be a continuous random variable on $(-\infty, \infty)$ with probability density function $p(x)$.

The differential entropy of $X$ is

$$H[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx.$$

We want to maximize $H[p]$ subject to the following constraints:

### Normalization

$$\int_{-\infty}^{\infty} p(x) \, dx = 1.$$

### Fixed mean

$$\int_{-\infty}^{\infty} x \, p(x) \, dx = \mu.$$

### Fixed variance

$$\int_{-\infty}^{\infty} (x - \mu)^2 p(x) \, dx = \sigma^2.$$

---

## 1. Construct the Lagrangian Functional

Introduce Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2$. Define

$$\begin{aligned}
\mathcal{L}[p]
={}&-\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx \\
&-\lambda_0 \left( \int_{-\infty}^{\infty} p(x) \, dx - 1 \right) \\
&-\lambda_1 \left( \int_{-\infty}^{\infty} x \, p(x) \, dx - \mu \right) \\
&-\lambda_2 \left( \int_{-\infty}^{\infty} (x - \mu)^2 p(x) \, dx - \sigma^2 \right).
\end{aligned}$$

To find the maximizing density, take the functional derivative with respect to $p(x)$ and set it equal to zero:

$$\frac{\delta \mathcal{L}}{\delta p(x)} = 0.$$

Since

$$\frac{\delta}{\delta p(x)} \left[-p(x) \ln p(x)\right] = -\ln p(x) - 1,$$

we obtain

$$-\ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0.$$

Therefore,

$$\ln p(x) = -1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2.$$

Exponentiating,

$$p(x) = \exp\left[-1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2\right].$$

Thus, the maximum-entropy density must have an exponential quadratic form.

Because the density must be centred at the prescribed mean $\mu$, the linear term can be absorbed into the quadratic expression. Hence, the density can be written as

$$p(x) = A \exp\left[-a(x - \mu)^2\right],$$

where $A > 0$ and $a > 0$ are constants.
## 2. Determine the Normalization Constant

The normalization condition requires

$$A \int_{-\infty}^{\infty} e^{-a(x - \mu)^2} \, dx = 1.$$

Using the Gaussian integral

$$\int_{-\infty}^{\infty} e^{-a(x - \mu)^2} \, dx = \sqrt{\frac{\pi}{a}},$$

we obtain

$$A \sqrt{\frac{\pi}{a}} = 1.$$

Therefore,

$$A = \sqrt{\frac{a}{\pi}}.$$

Hence,

$$p(x) = \sqrt{\frac{a}{\pi}} \exp\left[-a(x - \mu)^2\right].$$

---

## 3. Apply the Variance Constraint

For a density of the form

$$p(x) = \sqrt{\frac{a}{\pi}} e^{-a(x - \mu)^2},$$

the variance is

$$\mathbb{E}\left[(X - \mu)^2\right] = \frac{1}{2a}.$$

The required variance is $\sigma^2$, so

$$\frac{1}{2a} = \sigma^2.$$

Therefore,

$$a = \frac{1}{2\sigma^2}.$$

The normalization constant becomes

$$A = \sqrt{\frac{1}{2\pi\sigma^2}} = \frac{1}{\sqrt{2\pi\sigma^2}}.$$

Thus,

$$\boxed{p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left[-\frac{(x - \mu)^2}{2\sigma^2}\right]}.$$

Therefore,

$$\boxed{X \sim \mathcal{N}(\mu, \sigma^2)}.$$
## 4. Verification That It Is a Maximum

Let

$$\phi(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left[-\frac{(x - \mu)^2}{2\sigma^2}\right]$$

be the Gaussian density with mean $\mu$ and variance $\sigma^2$.

The Kullback–Leibler divergence between any density $p(x)$ satisfying the same constraints and $\phi(x)$ is

$$D_{\mathrm{KL}}(p \| \phi) = \int_{-\infty}^{\infty} p(x) \ln\frac{p(x)}{\phi(x)} \, dx \geq 0.$$

Expanding,

$$D_{\mathrm{KL}}(p \| \phi) = \int p(x) \ln p(x) \, dx - \int p(x) \ln \phi(x) \, dx.$$

Since

$$\ln \phi(x) = -\frac{1}{2}\ln(2\pi\sigma^2) - \frac{(x - \mu)^2}{2\sigma^2},$$

and $p(x)$ has variance $\sigma^2$,

$$\begin{aligned}
\int p(x) \ln \phi(x) \, dx
&= -\frac{1}{2}\ln(2\pi\sigma^2) - \frac{1}{2\sigma^2} \int (x - \mu)^2 p(x) \, dx \\
&= -\frac{1}{2}\ln(2\pi\sigma^2) - \frac{1}{2}.
\end{aligned}$$

Therefore,

$$D_{\mathrm{KL}}(p \| \phi) = -H[p] + \frac{1}{2}\ln(2\pi\sigma^2) + \frac{1}{2}.$$

Because $D_{\mathrm{KL}}(p \| \phi) \geq 0$,

$$H[p] \leq \frac{1}{2}\ln(2\pi\sigma^2) + \frac{1}{2}.$$

Hence,

$$\boxed{H[p] \leq \frac{1}{2}\ln(2\pi e\sigma^2)}.$$

Equality occurs only when

$$p(x) = \phi(x)$$

almost everywhere. Thus, the Gaussian distribution is the unique maximum-entropy distribution for a continuous random variable with a specified mean and variance.
## Final Result

Among all probability distributions on $(-\infty, \infty)$ having mean $\mu$ and variance $\sigma^2$, the one with maximum differential entropy is

$$\boxed{X \sim \mathcal{N}(\mu, \sigma^2)}$$

with density

$$\boxed{p_X(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)}.$$

Its maximum entropy is

$$\boxed{H(X) = \frac{1}{2}\ln(2\pi e\sigma^2)}.$$

# Maximum-Entropy Logistic Classification of Emails

Let

$$q(\mathbf x) = P(Y = 1 \mid \mathbf x)$$

denote the conditional probability that an email is routed to the **Technical** team. Therefore,

$$P(Y = 0 \mid \mathbf x) = 1 - q(\mathbf x)$$

is the probability that it is routed to **Finance**.

The feature states and their probabilities are

$$\begin{array}{c|c|c}
\text{Scenario} & \mathbf x & P(\mathbf x) \\
\hline
A & (1,0,0)^T & 0.40 \\
B & (0,1,0)^T & 0.25 \\
C & (0,0,1)^T & 0.05 \\
D & (0,1,1)^T & 0.15 \\
E & (1,1,0)^T & 0.05
\end{array}$$

These probabilities add to

$$0.40 + 0.25 + 0.05 + 0.15 + 0.05 = 0.90.$$

To form a complete probability distribution, assume that the remaining state is

$$\mathbf x_0 = (0,0,0)^T, \qquad P(\mathbf x_0) = 0.10.$$

Since $\mathbf x_0$ contains no active feature, it does not affect any of the three supplied moment constraints.

---

## 1. Conditional Maximum-Entropy Problem

The conditional entropy of $Y$ given $X$ is

$$H(Y \mid X) = -\sum_{\mathbf x} P(\mathbf x) \sum_{y=0}^{1} P(y \mid \mathbf x) \ln P(y \mid \mathbf x).$$

Using $q(\mathbf x) = P(Y = 1 \mid \mathbf x)$, this becomes

$$H(Y \mid X) = -\sum_{\mathbf x} P(\mathbf x) \left[ q(\mathbf x) \ln q(\mathbf x) + (1 - q(\mathbf x)) \ln(1 - q(\mathbf x)) \right].$$

We maximize this entropy subject to

$$\sum_{\mathbf x} P(\mathbf x) x_1 q(\mathbf x) = 0.05,$$

$$\sum_{\mathbf x} P(\mathbf x) x_2 q(\mathbf x) = 0.15,$$

and

$$\sum_{\mathbf x} P(\mathbf x) x_3 q(\mathbf x) = 0.16.$$

These are exactly the conditions

$$\mathbb E[x_i \mathbb I(Y = 1)] = \mathbb E[x_i P(Y = 1 \mid X)] = c_i,$$

where

$$\mathbf c = \begin{pmatrix} 0.05 \\ 0.15 \\ 0.16 \end{pmatrix}.$$
## 2. Lagrangian

Introduce Lagrange multipliers

$$\boldsymbol{\lambda} = \begin{pmatrix} \lambda_1 \\ \lambda_2 \\ \lambda_3 \end{pmatrix}.$$

The Lagrangian is

$$\begin{aligned}
\mathcal L
={}&H(Y \mid X) \\
&+ \sum_{i=1}^{3} \lambda_i \left[ \sum_{\mathbf x} P(\mathbf x) x_i q(\mathbf x) - c_i \right].
\end{aligned}$$

Differentiate with respect to $q(\mathbf x)$:

$$\frac{\partial \mathcal L}{\partial q(\mathbf x)} = P(\mathbf x) \left[ \ln\left(\frac{1 - q(\mathbf x)}{q(\mathbf x)}\right) + \boldsymbol{\lambda}^{T} \mathbf x \right].$$

At the maximum,

$$\frac{\partial \mathcal L}{\partial q(\mathbf x)} = 0.$$

Therefore,

$$\ln\left(\frac{1 - q(\mathbf x)}{q(\mathbf x)}\right) + \boldsymbol{\lambda}^{T} \mathbf x = 0.$$

Rearranging,

$$\ln\left(\frac{q(\mathbf x)}{1 - q(\mathbf x)}\right) = \boldsymbol{\lambda}^{T} \mathbf x.$$

The left side is the log-odds or logit function. Hence,

$$\boxed{q(\mathbf x) = P(Y = 1 \mid \mathbf x) = \frac{\exp(\boldsymbol{\lambda}^{T} \mathbf x)}{1 + \exp(\boldsymbol{\lambda}^{T} \mathbf x)}}$$

or

$$\boxed{P(Y = 1 \mid \mathbf x) = \sigma(\boldsymbol{\lambda}^{T} \mathbf x)}$$

where

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

is the logistic sigmoid function.

Therefore, logistic regression arises naturally as the maximum-entropy conditional distribution.

For $y \in \{0, 1\}$, the complete conditional distribution is

$$\boxed{P(Y = y \mid \mathbf x) = \frac{\exp\left(y \boldsymbol{\lambda}^{T} \mathbf x\right)}{1 + \exp\left(\boldsymbol{\lambda}^{T} \mathbf x\right)}}.$$
## 3. Probabilities for the Five Scenarios

Let

$$q_A = P(Y = 1 \mid \mathbf x_A),$$

and define $q_B, q_C, q_D, q_E$ similarly.

The logistic model gives

$$q_A = \sigma(\lambda_1),$$

$$q_B = \sigma(\lambda_2),$$

$$q_C = \sigma(\lambda_3),$$

$$q_D = \sigma(\lambda_2 + \lambda_3),$$

and

$$q_E = \sigma(\lambda_1 + \lambda_2).$$

For the zero-feature state,

$$q_0 = \sigma(0) = \frac{1}{2}.$$

This occurs because no overall class-probability constraint or intercept term has been supplied.

---

## 4. Moment Equations

### Constraint 1

Only Scenarios A and E have $x_1 = 1$. Therefore,

$$0.40 q_A + 0.05 q_E = 0.05.$$

Thus,

$$\boxed{0.40 \sigma(\lambda_1) + 0.05 \sigma(\lambda_1 + \lambda_2) = 0.05}.$$

### Constraint 2

Scenarios B, D and E have $x_2 = 1$. Therefore,

$$0.25 q_B + 0.15 q_D + 0.05 q_E = 0.15.$$

Thus,

$$\boxed{0.25 \sigma(\lambda_2) + 0.15 \sigma(\lambda_2 + \lambda_3) + 0.05 \sigma(\lambda_1 + \lambda_2) = 0.15}.$$

### Constraint 3

Scenarios C and D have $x_3 = 1$. Therefore,

$$0.05 q_C + 0.15 q_D = 0.16.$$

Thus,

$$\boxed{0.05 \sigma(\lambda_3) + 0.15 \sigma(\lambda_2 + \lambda_3) = 0.16}.$$

These nonlinear equations are solved numerically.
## 5. Logistic-Regression Parameters

The numerical solution is

$$\boxed{\lambda_1 \approx -1.9725}$$

$$\boxed{\lambda_2 \approx -1.7762}$$

$$\boxed{\lambda_3 \approx 2.8804}.$$

Therefore, the maximum-entropy classifier is

$$\boxed{P(Y = 1 \mid \mathbf x) = \frac{1}{1 + \exp\left[-\left(-1.9725x_1 - 1.7762x_2 + 2.8804x_3\right)\right]}}.$$

Equivalently,

$$\boxed{P(Y = 1 \mid \mathbf x) = \sigma\left(-1.9725x_1 - 1.7762x_2 + 2.8804x_3\right)}.$$

The Finance probability is

$$\boxed{P(Y = 0 \mid \mathbf x) = 1 - P(Y = 1 \mid \mathbf x)}.$$
## 6. Scenario Probabilities

Substituting each feature vector gives:

$$\begin{array}{c|c|c|c}
\text{Scenario} & P(Y=1 \mid \mathbf x) & P(Y=0 \mid \mathbf x) & \text{Predicted class} \\
\hline
A:\ (1,0,0) & 0.1221 & 0.8779 & \text{Finance} \\
B:\ (0,1,0) & 0.1448 & 0.8552 & \text{Finance} \\
C:\ (0,0,1) & 0.9469 & 0.0531 & \text{Technical} \\
D:\ (0,1,1) & 0.7510 & 0.2490 & \text{Technical} \\
E:\ (1,1,0) & 0.0230 & 0.9770 & \text{Finance} \\
\mathbf x_0:\ (0,0,0) & 0.5000 & 0.5000 & \text{Undetermined}
\end{array}$$

Using a classification threshold of $0.5$,

$$\widehat Y = \begin{cases}
1, & P(Y = 1 \mid \mathbf x) > 0.5, \\
0, & P(Y = 1 \mid \mathbf x) < 0.5.
\end{cases}$$

Thus:

* Pure Financial emails are classified as Finance.
* Pure Access emails are classified as Finance.
* Pure Technical emails are classified as Technical.
* Access–Technical emails are classified as Technical.
* Billing–Access emails are classified strongly as Finance.
## 7. Verification of the Constraints

### Financial-feature constraint

$$\begin{aligned}
\mathbb E[x_1 \mathbb I(Y = 1)]
&= 0.40 q_A + 0.05 q_E \\
&= 0.40(0.1221) + 0.05(0.0230) \\
&\approx 0.0500.
\end{aligned}$$

Therefore,

$$\boxed{\mathbb E[x_1 \mathbb I(Y = 1)] = 0.05}.$$

### Account-access constraint

$$\begin{aligned}
\mathbb E[x_2 \mathbb I(Y = 1)]
={}&0.25 q_B + 0.15 q_D + 0.05 q_E \\
={}&0.25(0.1448) + 0.15(0.7510) + 0.05(0.0230) \\
\approx{}&0.1500.
\end{aligned}$$

Therefore,

$$\boxed{\mathbb E[x_2 \mathbb I(Y = 1)] = 0.15}.$$

### Technical-feature constraint

$$\begin{aligned}
\mathbb E[x_3 \mathbb I(Y = 1)]
&= 0.05 q_C + 0.15 q_D \\
&= 0.05(0.9469) + 0.15(0.7510) \\
&\approx 0.1600.
\end{aligned}$$

Therefore,

$$\boxed{\mathbb E[x_3 \mathbb I(Y = 1)] = 0.16}.$$

All three empirical constraints are satisfied.
## 8. Interpretation of the Coefficients

The fitted log-odds are

$$\ln \frac{P(Y = 1 \mid \mathbf x)}{P(Y = 0 \mid \mathbf x)} = -1.9725x_1 - 1.7762x_2 + 2.8804x_3.$$

The coefficient of $x_1$ is negative:

$$\lambda_1 = -1.9725.$$

Therefore, detecting financial context strongly decreases the probability of routing the email to Technical support.

The coefficient of $x_2$ is also negative:

$$\lambda_2 = -1.7762.$$

An account-access feature by itself favours Finance under the supplied constraints. However, when it appears together with a technical feature, the positive contribution of $x_3$ dominates.

The coefficient of $x_3$ is strongly positive:

$$\lambda_3 = 2.8804.$$

Thus, detecting a bug or malfunction substantially increases the probability of routing the email to Technical support.

For Scenario D,

$$\lambda_2 + \lambda_3 = -1.7762 + 2.8804 = 1.1042 > 0,$$

so

$$P(Y = 1 \mid \mathbf x_D) = \sigma(1.1042) \approx 0.7510.$$

For Scenario E,

$$\lambda_1 + \lambda_2 = -1.9725 - 1.7762 = -3.7487,$$

so

$$P(Y = 1 \mid \mathbf x_E) = \sigma(-3.7487) \approx 0.0230.$$

This means an email containing both financial and access features is classified very strongly as Finance.
## Final Answer

The maximum-entropy conditional distribution consistent with the supplied feature expectations is the logistic-regression model

$$\boxed{P(Y = 1 \mid \mathbf x) = \sigma\left(-1.9725x_1 - 1.7762x_2 + 2.8804x_3\right)}$$

and

$$\boxed{P(Y = 0 \mid \mathbf x) = 1 - \sigma\left(-1.9725x_1 - 1.7762x_2 + 2.8804x_3\right)}.$$

The resulting Technical-routing probabilities are

$$\boxed{q_A = 0.1221, \quad q_B = 0.1448, \quad q_C = 0.9469, \quad q_D = 0.7510, \quad q_E = 0.0230}.$$

Therefore, Scenarios A, B and E are classified as Finance, while Scenarios C and D are classified as Technical.

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



# Maximum-Entropy Risk-Neutral Distribution of Asset X

Let

$$S = S_T > 0$$

denote the price of Asset X after one month, and let $p(s)$ be its risk-neutral probability density.

The supplied constraints are

$$\int_0^\infty p(s) \, ds = 1,$$

$$\int_0^\infty s \, p(s) \, ds = 102,$$

$$\int_0^\infty s^2 p(s) \, ds = 10600,$$

and

$$\int_0^\infty \ln(s) \, p(s) \, ds = 4.60.$$

Strictly, since the logarithm requires a dimensionless argument, the final constraint may be interpreted as

$$\mathbb E\left[\ln\left(\frac{S_T}{\$1}\right)\right] = 4.60.$$

---

## 1. Check the Market Constraints

The variance implied by the first two moments is

$$\operatorname{Var}(S_T) = \mathbb E[S_T^2] - \mathbb E[S_T]^2.$$

Therefore,

$$\operatorname{Var}(S_T) = 10600 - (102)^2.$$

Since

$$102^2 = 10404,$$

we obtain

$$\boxed{\operatorname{Var}(S_T) = 196}.$$

Hence, the standard deviation is

$$\boxed{\sigma_S = 14}.$$

The logarithmic constraint is consistent with Jensen's inequality because

$$\mathbb E[\ln S_T] \leq \ln \mathbb E[S_T].$$

Here,

$$4.60 < \ln(102) \approx 4.62497.$$

Thus, the supplied logarithmic moment is feasible and represents downside skew relative to a tightly concentrated distribution around $102$.
## 2. Maximum-Entropy Objective

The differential entropy of $S_T$ is

$$H[p] = -\int_0^\infty p(s) \ln p(s) \, ds.$$

We seek the density $p(s)$ that solves

$$\boxed{\max_{p} \left[ -\int_0^\infty p(s) \ln p(s) \, ds \right]}$$

subject to

$$\int_0^\infty p(s) \, ds = 1,$$

$$\int_0^\infty s \, p(s) \, ds = 102,$$

$$\int_0^\infty s^2 p(s) \, ds = 10600,$$

and

$$\int_0^\infty \ln(s) p(s) \, ds = 4.60.$$

These constraints are linear in the unknown density $p(s)$.
## 3. Construct the Lagrangian Functional

Introduce Lagrange multipliers

$$\lambda_0, \lambda_1, \lambda_2, \lambda_3.$$

The Lagrangian is

$$\begin{aligned}
\mathcal L[p]
={}& -\int_0^\infty p(s) \ln p(s) \, ds \\
&+ \lambda_0 \left[ \int_0^\infty p(s) \, ds - 1 \right] \\
&+ \lambda_1 \left[ \int_0^\infty s p(s) \, ds - 102 \right] \\
&+ \lambda_2 \left[ \int_0^\infty s^2 p(s) \, ds - 10600 \right] \\
&+ \lambda_3 \left[ \int_0^\infty \ln(s) p(s) \, ds - 4.60 \right].
\end{aligned}$$

To determine the maximizing density, take the functional derivative with respect to $p(s)$:

$$\frac{\delta \mathcal L}{\delta p(s)} = -\ln p(s) - 1 + \lambda_0 + \lambda_1 s + \lambda_2 s^2 + \lambda_3 \ln s.$$

At the optimum,

$$\frac{\delta \mathcal L}{\delta p(s)} = 0.$$

Therefore,

$$-\ln p(s) - 1 + \lambda_0 + \lambda_1 s + \lambda_2 s^2 + \lambda_3 \ln s = 0.$$

Rearranging,

$$\ln p(s) = \lambda_0 - 1 + \lambda_1 s + \lambda_2 s^2 + \lambda_3 \ln s.$$

Exponentiating,

$$p(s) = e^{\lambda_0 - 1} s^{\lambda_3} e^{\lambda_1 s + \lambda_2 s^2}.$$

Introducing the normalizing partition function $Z$,

$$Z(\lambda_1, \lambda_2, \lambda_3) = \int_0^\infty s^{\lambda_3} \exp\left(\lambda_1 s + \lambda_2 s^2\right) \, ds,$$

the maximum-entropy density has the form

$$\boxed{p^*(s) = \frac{1}{Z} s^{\lambda_3} \exp\left(\lambda_1 s + \lambda_2 s^2\right), \qquad s > 0}.$$

For this integral to converge, the parameters must satisfy

$$\boxed{\lambda_2 < 0}$$

so that the quadratic term controls the upper tail, and

$$\boxed{\lambda_3 > -1}$$

so that the density is integrable near $s = 0$.
## 4. Determining the Lagrange Multipliers

The partition function is

$$Z = \int_0^\infty s^{\lambda_3} e^{\lambda_1 s + \lambda_2 s^2} \, ds.$$

The required moments are obtained by differentiating $\ln Z$:

$$\frac{\partial \ln Z}{\partial \lambda_1} = \mathbb E[S_T],$$

$$\frac{\partial \ln Z}{\partial \lambda_2} = \mathbb E[S_T^2],$$

and

$$\frac{\partial \ln Z}{\partial \lambda_3} = \mathbb E[\ln S_T].$$

Thus, the multipliers solve

$$\boxed{\frac{\partial \ln Z}{\partial \lambda_1} = 102}$$

$$\boxed{\frac{\partial \ln Z}{\partial \lambda_2} = 10600}$$

and

$$\boxed{\frac{\partial \ln Z}{\partial \lambda_3} = 4.60}.$$

These equations do not have a convenient elementary closed-form solution, so they are solved numerically.

The numerical solution is

$$\boxed{\lambda_1 \approx 0.54084212}$$

$$\boxed{\lambda_2 \approx -0.0026021654}$$

and

$$\boxed{\lambda_3 \approx -0.99998904}.$$

The corresponding logarithm of the partition function is

$$\boxed{\ln Z \approx 27.025724}.$$

Therefore,

$$Z \approx e^{27.025724} \approx 5.4591 \times 10^{11}.$$

Hence,

$$\frac{1}{Z} \approx 1.8318 \times 10^{-12}.$$
## 5. Final Maximum-Entropy Density

Substituting the numerical values gives

$$\boxed{p^*(s) = 1.8318 \times 10^{-12} \, s^{-0.99998904} \exp\left(0.54084212 s - 0.0026021654 s^2\right), \qquad s > 0}.$$

Equivalently,

$$\boxed{p^*(S_T) = \frac{S_T^{-0.99998904} \exp\left(0.54084212 S_T - 0.0026021654 S_T^2\right)}{5.4591 \times 10^{11}}, \qquad S_T > 0}.$$

This is the least-biased continuous density satisfying all three market constraints.

---

## 6. Verification of the Constraints

The density is normalized:

$$\int_0^\infty p^*(s) \, ds = 1.$$

Numerically, it satisfies

$$\boxed{\int_0^\infty s p^*(s) \, ds = 102}$$

$$\boxed{\int_0^\infty s^2 p^*(s) \, ds = 10600}$$

and

$$\boxed{\int_0^\infty \ln(s) p^*(s) \, ds = 4.60}.$$

Therefore,

$$\mathbb E[S_T] = 102,$$

$$\operatorname{Var}(S_T) = 10600 - 102^2 = 196,$$

and

$$\mathbb E[\ln S_T] = 4.60.$$
## 7. Interpretation of the Distribution

The ordinary mean and variance constraints create a main concentration of probability near the forward price. The principal interior peak occurs close to

$$\boxed{S_T \approx 102.04}.$$

However, the logarithmic constraint forces the distribution to include additional downside risk.

Since

$$\lambda_3 \approx -0.999989,$$

the density behaves near zero approximately as

$$p^*(s) \propto s^{-0.999989}.$$

Thus, the density becomes very large as $s \to 0^+$, although it remains integrable because

$$\lambda_3 > -1.$$

This produces an extremely thin but very long downside component near zero. Even a very small amount of probability near zero has a substantial influence on

$$\mathbb E[\ln S_T],$$

because

$$\ln s \to -\infty \qquad \text{as } s \to 0^+.$$

Therefore, the logarithmic constraint captures crash risk that would not be visible from the mean and variance alone.
## 8. Why the Distribution Is Not Gaussian

If only the mean and variance were imposed on a variable over the entire real line, the maximum-entropy distribution would be Gaussian.

Here, however:

1. $S_T$ is restricted to positive values.
2. A logarithmic-moment constraint is imposed.
3. The logarithmic constraint introduces asymmetry and downside sensitivity.

The resulting density is therefore

$$p^*(s) \propto s^{\lambda_3} e^{\lambda_1 s + \lambda_2 s^2},$$

rather than a normal density.

This distribution can represent asymmetric crash risk while preserving the market-implied forward price and variance.

---

## 9. Maximum Entropy Value

For the optimal density,

$$\ln p^*(s) = -\ln Z + \lambda_1 s + \lambda_2 s^2 + \lambda_3 \ln s.$$

Therefore,

$$\begin{aligned}
H[p^*]
&= -\mathbb E[\ln p^*(S_T)] \\
&= \ln Z - \lambda_1 \mathbb E[S_T] - \lambda_2 \mathbb E[S_T^2] - \lambda_3 \mathbb E[\ln S_T].
\end{aligned}$$

Substituting the values,

$$H[p^*] = 27.025724 - (0.54084212)(102) - (-0.0026021654)(10600) - (-0.99998904)(4.60).$$

This gives

$$\boxed{H[p^*] \approx 4.04273}.$$

Because entropy is strictly concave in $p$, this feasible stationary density is the unique maximum-entropy solution.
# Final Answer

The maximum-entropy problem is

$$\max_p \left[ -\int_0^\infty p(s) \ln p(s) \, ds \right]$$

subject to

$$\int_0^\infty p(s) \, ds = 1,$$

$$\int_0^\infty s p(s) \, ds = 102,$$

$$\int_0^\infty s^2 p(s) \, ds = 10600,$$

and

$$\int_0^\infty \ln(s) p(s) \, ds = 4.60.$$

Its general solution is

$$\boxed{p^*(s) = \frac{1}{Z} s^{\lambda_3} e^{\lambda_1 s + \lambda_2 s^2}, \qquad s > 0}$$

with numerical multipliers

$$\boxed{\lambda_1 = 0.54084212, \qquad \lambda_2 = -0.0026021654, \qquad \lambda_3 = -0.99998904}.$$

Thus,

$$\boxed{p^*(s) = 1.8318 \times 10^{-12} s^{-0.99998904} \exp\left(0.54084212 s - 0.0026021654 s^2\right), \qquad s > 0}.$$

This density has mean $102$, variance $196$, and logarithmic expectation $4.60$. It has a main peak near $102$, together with an integrable near-zero component representing the strong market-implied downside risk.
